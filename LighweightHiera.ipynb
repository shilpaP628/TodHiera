```python
# ============================================================
# Toddler Activity Recognition - Modified Hiera
# Reduced stages=(2,3,10,3) + Partial Pretrained Loading + FULL TRAINING
# + Class weights inside each fold
# + Consistent Video Augmentations
# + LR logging + Top-2 Accuracy + Edge Deployment Metrics
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

!pip install -q torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
!pip install -q hiera-transformer av opencv-python tqdm scikit-learn seaborn pandas openpyxl thop

import os
import random
import time
from pathlib import Path
from collections import Counter, OrderedDict

import av
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.transforms import functional as TF
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    precision_score, recall_score, roc_auc_score, top_k_accuracy_score
)
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns
import hiera
from thop import profile, clever_format

# ────────────────────────────────────────────────
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    torch.backends.cudnn.benchmark = True

# ===================== Settings =====================
IMAGE_HEIGHT, IMAGE_WIDTH = 224, 224
SEQUENCE_LENGTH = 16
BATCH_SIZE = 4
EPOCHS = 25
LR_BACKBONE = 1e-5
LR_HEAD = 5e-5
WEIGHT_DECAY = 0.05
NUM_FOLDS = 5
NUM_WORKERS = 4
PATIENCE = 4
PIN_MEMORY = True
USE_AMP = True
TRAIN_LAST_PERCENT = 1.0          # FULL TRAINING (all blocks + head)

DATA_DIR = "/content/drive/MyDrive/DatasetFinal"
SAVE_DIR = "hiera_modified_stages23103_full_train"
os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 16 Classes =====================
CLASSES_LIST = [
    'Climb_Down_Stairs', 'Climb_Up_Stairs', 'Crawl', 'Drink', 'Eat',
    'Emesis', 'Exit_Crib', 'Fall_Down', 'Hold_Knife', 'Play_Slide',
    'Play_Stacking_Rings', 'Play_Swing', 'Play_Walker', 'Sleep', 'W_Sit', 'Walk'
]
NUM_CLASSES = len(CLASSES_LIST)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 70)
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
print(f"Strategy: Reduced Hiera (2,3,10,3) + Partial Pretrained + FULL TRAINING")
print("=" * 70)


# ============================================================
# VIDEO AUGMENTATION
# ============================================================
class VideoAugmentation:
    def __init__(self, size=224, is_train=True):
        self.size = size
        self.is_train = is_train
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

    def __call__(self, frames):
        pil_frames = [transforms.ToPILImage()(f) for f in frames]

        if self.is_train:
            if random.random() < 0.5:
                pil_frames = [TF.hflip(f) for f in pil_frames]

            brightness = random.uniform(0.8, 1.2)
            contrast   = random.uniform(0.8, 1.2)
            saturation = random.uniform(0.8, 1.2)
            hue        = random.uniform(-0.05, 0.05)

            pil_frames = [TF.adjust_brightness(f, brightness) for f in pil_frames]
            pil_frames = [TF.adjust_contrast(f, contrast) for f in pil_frames]
            pil_frames = [TF.adjust_saturation(f, saturation) for f in pil_frames]
            pil_frames = [TF.adjust_hue(f, hue) for f in pil_frames]

        tensors = []
        for f in pil_frames:
            f = TF.resize(f, [self.size, self.size])
            t = TF.to_tensor(f)
            if self.is_train and random.random() < 0.3:
                noise = torch.randn_like(t) * 0.02
                t = torch.clamp(t + noise, 0.0, 1.0)
            t = self.normalize(t)
            tensors.append(t)

        return torch.stack(tensors)


def load_video_frames_raw(video_path, num_frames=16):
    container = av.open(video_path)
    frames = []
    for frame in container.decode(video=0):
        img = frame.to_ndarray(format="rgb24")
        frames.append(img)
    container.close()

    if len(frames) == 0:
        return None

    indices = np.linspace(0, len(frames) - 1, num_frames).astype(int)
    return [frames[i] for i in indices]


# ============================================================
# DATASET
# ============================================================
class ActivityDataset(Dataset):
    def __init__(self, data_dir, classes_list, num_frames=16, size=224, is_train=True):
        self.samples = []
        self.labels = []
        self.num_frames = num_frames
        self.size = size
        self.is_train = is_train
        self.class_to_idx = {cls: idx for idx, cls in enumerate(classes_list)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}
        self.augment = VideoAugmentation(size=size, is_train=is_train)

        for class_name in classes_list:
            class_path = Path(data_dir) / class_name
            if not class_path.is_dir():
                print(f"Warning: folder not found → {class_path}")
                continue
            count = 0
            for video_file in sorted(class_path.glob("*")):
                if video_file.suffix.lower() in {".mp4", ".avi", ".mov", ".mkv"}:
                    self.samples.append(str(video_file))
                    self.labels.append(self.class_to_idx[class_name])
                    count += 1
            print(f"Loaded {count} videos from '{class_name}'")

        self.labels = np.array(self.labels)
        print(f"\nTotal videos: {len(self.samples)}")
        print("Class distribution:", dict(Counter(self.labels)))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path = self.samples[idx]
        label = self.labels[idx]
        frames = load_video_frames_raw(path, self.num_frames)

        if frames is None:
            frames = [np.zeros((self.size, self.size, 3), dtype=np.uint8) for _ in range(self.num_frames)]

        video_tensor = self.augment(frames)
        return video_tensor, label, path


def collate_fn(batch):
    videos, labels, paths = zip(*batch)
    videos = torch.stack(videos)
    labels = torch.tensor(labels, dtype=torch.long)
    return videos, labels, list(paths)


# ============================================================
# FULL TRAINING (no freezing)
# ============================================================
def freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT):
    """
    Full training mode: every parameter is trainable.
    The train_last_percent argument is kept only for API compatibility.
    """
    for param in model.parameters():
        param.requires_grad = True

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"\n[Full Training] Trainable params: {trainable:,} / {total:,} "
          f"({100 * trainable / total:.1f}%)")


# ============================================================
# PARTIAL WEIGHT LOADING + REDUCED MODEL
# ============================================================
def create_reduced_hiera_with_partial_pretrained(
    num_classes=16,
    stages=(2, 3, 10, 3),
    embed_dim=96,
    device="cuda"
):
    model = hiera.Hiera(
        input_size=(16, 224, 224),
        in_chans=3,
        embed_dim=embed_dim,
        num_heads=1,
        num_classes=num_classes,
        stages=stages,
        q_pool=3,
        q_stride=(1, 2, 2),
        mask_unit_size=(1, 8, 8),
        patch_kernel=(3, 7, 7),
        patch_stride=(2, 4, 4),
        patch_padding=(1, 3, 3),
        sep_pos_embed=True,
        drop_path_rate=0.15,
    )

    print("\nLoading original pretrained Hiera-Base weights...")
    pretrained = hiera.hiera_base_16x224(
        pretrained=True,
        checkpoint="mae_k400_ft_k400",
        num_classes=400
    )
    pretrained_dict = pretrained.state_dict()
    model_dict = model.state_dict()

    new_state_dict = OrderedDict()
    loaded_keys = []
    skipped_keys = []

    for key, value in pretrained_dict.items():
        if "head" in key:
            skipped_keys.append(key)
            continue

        if "blocks" in key:
            parts = key.split(".")
            block_idx = int(parts[1])

            if block_idx <= 4:
                new_block_idx = block_idx
            elif 5 <= block_idx <= 14:
                new_block_idx = block_idx
            elif block_idx >= 21:
                new_block_idx = block_idx - 6
            else:
                skipped_keys.append(key)
                continue

            new_key = key.replace(f"blocks.{block_idx}.", f"blocks.{new_block_idx}.")

            if new_key in model_dict and model_dict[new_key].shape == value.shape:
                new_state_dict[new_key] = value
                loaded_keys.append(new_key)
            else:
                skipped_keys.append(key)
        else:
            if key in model_dict and model_dict[key].shape == value.shape:
                new_state_dict[key] = value
                loaded_keys.append(key)
            else:
                skipped_keys.append(key)

    missing, unexpected = model.load_state_dict(new_state_dict, strict=False)

    print("\n" + "="*65)
    print("PARTIAL WEIGHT LOADING SUMMARY")
    print("="*65)
    print(f"Successfully loaded keys : {len(loaded_keys)}")
    print(f"Skipped keys             : {len(skipped_keys)}")
    print(f"Missing keys             : {len(missing)}")
    print(f"Unexpected keys          : {len(unexpected)}")
    print("="*65)

    return model.to(device)


def create_hiera_model():
    model = create_reduced_hiera_with_partial_pretrained(
        num_classes=NUM_CLASSES,
        stages=(2, 3, 10, 3),
        embed_dim=96,
        device=DEVICE
    )
    freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT)
    return model


# ============================================================
# EDGE DEPLOYMENT METRICS
# ============================================================
def print_edge_deployment_metrics(model, device=DEVICE, input_shape=(1, 3, 16, 224, 224)):
    print("\n" + "=" * 70)
    print("EDGE DEPLOYMENT METRICS")
    print("=" * 70)

    model.eval()
    model.to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    param_size_mb = total_params * 4 / (1024 ** 2)
    param_size_mb_fp16 = total_params * 2 / (1024 ** 2)

    print(f"\n[1] MEMORY FOOTPRINT")
    print(f"    Total parameters      : {total_params:,}")
    print(f"    Trainable parameters  : {trainable_params:,}")
    print(f"    Model size (FP32)     : {param_size_mb:.2f} MB")
    print(f"    Model size (FP16 est.): {param_size_mb_fp16:.2f} MB")

    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
        torch.cuda.empty_cache()
        dummy = torch.randn(*input_shape, device=device)
        with torch.no_grad():
            _ = model(dummy)
        peak_mem_mb = torch.cuda.max_memory_allocated() / (1024 ** 2)
        print(f"    Peak GPU memory (fwd) : {peak_mem_mb:.2f} MB")

    print(f"\n[2] COMPUTATIONAL COMPLEXITY (FLOPs)")
    try:
        dummy = torch.randn(*input_shape, device=device)
        flops, params = profile(model, inputs=(dummy,), verbose=False)
        flops_str, params_str = clever_format([flops, params], "%.3f")
        print(f"    FLOPs                 : {flops_str}")
        print(f"    Params (thop)         : {params_str}")
    except Exception as e:
        print(f"    FLOPs calculation failed: {e}")

    print(f"\n[3] INFERENCE TIME (Latency)")
    dummy = torch.randn(*input_shape, device=device)
    with torch.no_grad():
        for _ in range(10):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()

    n_runs = 50
    start = time.perf_counter()
    with torch.no_grad():
        for _ in range(n_runs):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    end = time.perf_counter()

    avg_latency_ms = (end - start) / n_runs * 1000
    fps = 1000.0 / avg_latency_ms
    print(f"    Avg latency (batch=1) : {avg_latency_ms:.2f} ms")
    print(f"    Throughput            : {fps:.1f} FPS")

    print(f"\n[4] POWER DRAW")
    print(f"    Measure on target edge device (Jetson / mobile NPU)")

    print(f"\n[5] THERMAL LIMITS")
    print(f"    Monitor on target hardware under sustained load")

    print(f"\n[6] MODEL ACCURACY")
    print(f"    See FINAL RESULTS section after cross-validation")
    print("=" * 70 + "\n")


# ============================================================
# LOAD FULL DATASET
# ============================================================
print("\nLoading full dataset...")
full_dataset_for_split = ActivityDataset(
    DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
)


# ============================================================
# TRAIN / EVAL
# ============================================================
def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for videos, labels, _ in tqdm(loader, desc=" train", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * videos.size(0)
        preds = logits.argmax(1)
        correct += (preds == labels).sum().item()
        total += videos.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    all_preds, all_labels, all_probs, all_paths = [], [], [], []

    for videos, labels, paths in tqdm(loader, desc=" val", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        total_loss += loss.item() * videos.size(0)
        probs = torch.softmax(logits.float(), dim=1)
        preds = logits.argmax(1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
        all_paths.extend(paths)

    return (total_loss / len(loader.dataset),
            np.array(all_preds),
            np.array(all_labels),
            np.array(all_probs),
            all_paths)


def plot_cm(cm, title, normalize=True):
    if normalize:
        cm_plot = cm.astype("float") / (cm.sum(axis=1)[:, np.newaxis] + 1e-8)
        fmt = ".2f"
    else:
        cm_plot = cm
        fmt = "d"

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm_plot, annot=True, fmt=fmt, cmap="Blues",
                xticklabels=CLASSES_LIST, yticklabels=CLASSES_LIST)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# PROFILE MODEL ONCE
# ============================================================
print("\nCreating temporary model for edge metrics...")
_temp_model = create_hiera_model()
print_edge_deployment_metrics(_temp_model)
del _temp_model
torch.cuda.empty_cache()


# ============================================================
# 5-FOLD CROSS VALIDATION
# ============================================================
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)

accuracy_scores = []
top2_scores = []
f1_macro_scores = []
f1_weighted_scores = []
precision_scores = []
recall_scores = []
auc_scores = []
confusion_matrices = []

overall_best_f1 = 0.0
overall_best_path = None

fold = 1
for train_idx, val_idx in skf.split(np.zeros(len(full_dataset_for_split)), full_dataset_for_split.labels):
    print(f"\n{'='*70}")
    print(f"========== Fold {fold}/{NUM_FOLDS} ==========")
    print(f"{'='*70}")

    train_dataset = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=True)
    val_dataset   = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False)

    train_subset = Subset(train_dataset, train_idx)
    val_subset   = Subset(val_dataset, val_idx)

    train_labels = full_dataset_for_split.labels[train_idx]
    class_weights = compute_class_weight("balanced", classes=np.unique(train_labels), y=train_labels)
    class_weight_tensor = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
    print("Class weights (this fold):", dict(enumerate(np.round(class_weights, 3))))

    train_loader = DataLoader(
        train_subset, batch_size=BATCH_SIZE, shuffle=True,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )
    val_loader = DataLoader(
        val_subset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )

    model = create_hiera_model()
    criterion = nn.CrossEntropyLoss(weight=class_weight_tensor)

    backbone_params = [p for n, p in model.named_parameters() if "head" not in n and p.requires_grad]
    head_params     = [p for n, p in model.named_parameters() if "head" in n and p.requires_grad]

    optimizer = torch.optim.AdamW([
        {"params": backbone_params, "lr": LR_BACKBONE},
        {"params": head_params,     "lr": LR_HEAD}
    ], weight_decay=WEIGHT_DECAY)

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    best_f1 = 0.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        val_loss, y_pred, y_true, y_prob, _ = evaluate(model, val_loader, criterion)
        scheduler.step()

        current_lrs = [group['lr'] for group in optimizer.param_groups]
        print(f"  → LRs: backbone={current_lrs[0]:.2e} | head={current_lrs[1]:.2e}")

        val_acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average="macro", zero_division=0)

        try:
            top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
        except Exception:
            top2_acc = 0.0

        print(f"Epoch {epoch+1:02d} | "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} "
              f"top2={top2_acc:.4f} macro_F1={f1_macro:.4f}")

        if f1_macro > best_f1:
            best_f1 = f1_macro
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
            print(f" → New best macro F1: {best_f1:.4f}")
        else:
            epochs_no_improve += 1
            print(f" → No improvement for {epochs_no_improve}/{PATIENCE} epochs")

        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch+1}")
            break

    model.load_state_dict(best_state)
    model.to(DEVICE)
    val_loss, y_pred, y_true, y_prob, video_paths = evaluate(model, val_loader, criterion)

    acc = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall_macro    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_macro        = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_weighted     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    try:
        top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
    except Exception:
        top2_acc = 0.0

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc_macro = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except Exception as e:
        print(f"AUC calculation warning: {e}")
        auc_macro = 0.0

    accuracy_scores.append(acc)
    top2_scores.append(top2_acc)
    f1_macro_scores.append(f1_macro)
    f1_weighted_scores.append(f1_weighted)
    precision_scores.append(precision_macro)
    recall_scores.append(recall_macro)
    auc_scores.append(auc_macro)

    cm = confusion_matrix(y_true, y_pred)
    confusion_matrices.append(cm)
    plot_cm(cm, f"Activity Recognition CM – Fold {fold}")

    print(f"\n{'='*60}")
    print(f"Fold {fold} Final Metrics:")
    print(f"{'='*60}")
    print(f"Accuracy        : {acc:.4f}")
    print(f"Top-2 Accuracy  : {top2_acc:.4f}")
    print(f"Precision (Macro): {precision_macro:.4f}")
    print(f"Recall (Macro)  : {recall_macro:.4f}")
    print(f"F1-Score (Macro): {f1_macro:.4f}")
    print(f"F1-Score (Weighted): {f1_weighted:.4f}")
    print(f"AUC (Macro OVR) : {auc_macro:.4f}")
    print(f"{'='*60}")

    print(f"\nFold {fold} Classification Report:")
    print(classification_report(y_true, y_pred, target_names=CLASSES_LIST, digits=4, zero_division=0))

    true_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_true]
    pred_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_pred]

    df = pd.DataFrame({
        "Video_Path": video_paths,
        "True_Class": true_class_names,
        "Predicted_Class": pred_class_names,
        "Correct": [t == p for t, p in zip(true_class_names, pred_class_names)]
    })

    excel_path = os.path.join(SAVE_DIR, f"predictions_fold_{fold}.xlsx")
    df.to_excel(excel_path, index=False)
    print(f"Excel saved → {excel_path}")

    fold_save_path = os.path.join(SAVE_DIR, f"best_hiera_modified_fold_{fold}.pth")
    torch.save({
        "model_state_dict": best_state,
        "fold": fold,
        "macro_f1": best_f1,
        "accuracy": acc,
        "top2_accuracy": top2_acc,
        "class_to_idx": full_dataset_for_split.class_to_idx,
        "idx_to_class": full_dataset_for_split.idx_to_class,
        "classes": CLASSES_LIST,
        "stages": (2, 3, 10, 3),
    }, fold_save_path)
    print(f"Model saved → {fold_save_path}")

    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1
        overall_best_path = fold_save_path

    del optimizer, scaler, model
    torch.cuda.empty_cache()
    fold += 1


# ============================================================
# FINAL RESULTS
# ============================================================
print("\n" + "="*70)
print("FINAL RESULTS (Modified Hiera - stages=(2,3,10,3) + Partial Pretrained + FULL TRAINING)")
print("="*70)

print(f"\nPer-fold Accuracy : {[f'{x:.4f}' for x in accuracy_scores]}")
print(f"Mean ± std Accuracy : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")

print(f"\nPer-fold Top-2 Accuracy : {[f'{x:.4f}' for x in top2_scores]}")
print(f"Mean ± std Top-2 : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")

print(f"\nPer-fold Precision : {[f'{x:.4f}' for x in precision_scores]}")
print(f"Mean ± std Precision : {np.mean(precision_scores):.4f} ± {np.std(precision_scores):.4f}")

print(f"\nPer-fold Recall : {[f'{x:.4f}' for x in recall_scores]}")
print(f"Mean ± std Recall : {np.mean(recall_scores):.4f} ± {np.std(recall_scores):.4f}")

print(f"\nPer-fold Macro F1 : {[f'{x:.4f}' for x in f1_macro_scores]}")
print(f"Mean ± std Macro F1 : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")

print(f"\nPer-fold Weighted F1 : {[f'{x:.4f}' for x in f1_weighted_scores]}")
print(f"Mean ± std Weighted F1 : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")

print(f"\nPer-fold AUC (Macro) : {[f'{x:.4f}' for x in auc_scores]}")
print(f"Mean ± std AUC : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")

conf_sum = np.sum(confusion_matrices, axis=0)
plot_cm(conf_sum, "Aggregated Confusion Matrix (All Folds)", normalize=False)
plot_cm(conf_sum, "Aggregated Confusion Matrix (Normalized)", normalize=True)

print(f"\nOverall best model (Macro F1={overall_best_f1:.4f}) saved at:")
print(overall_best_path)

print("\n" + "=" * 70)
print("EDGE DEPLOYMENT SUMMARY (with measured accuracy)")
print("=" * 70)
print(f"Model Architecture     : Modified Hiera stages=(2,3,10,3)")
print(f"Pretrained Loading     : Partial (from original Hiera-Base)")
print(f"Trainable Strategy     : FULL TRAINING (all blocks + Head)")
print(f"Mean Accuracy          : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")
print(f"Mean Top-2 Accuracy    : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")
print(f"Mean Macro F1          : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")
print(f"Mean Weighted F1       : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")
print(f"Mean AUC               : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("=" * 70)
```

```python
# ============================================================
# Toddler Activity Recognition - Modified Hiera
# Reduced stages=(2,3,10,3) + Partial Pretrained Loading + Last 50% Trainable
# + Class weights inside each fold
# + Consistent Video Augmentations
# + LR logging + Top-2 Accuracy + Edge Deployment Metrics
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

!pip install -q torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
!pip install -q hiera-transformer av opencv-python tqdm scikit-learn seaborn pandas openpyxl thop

import os
import random
import time
from pathlib import Path
from collections import Counter, OrderedDict

import av
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.transforms import functional as TF
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    precision_score, recall_score, roc_auc_score, top_k_accuracy_score
)
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns
import hiera
from thop import profile, clever_format

# ────────────────────────────────────────────────
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    torch.backends.cudnn.benchmark = True

# ===================== Settings =====================
IMAGE_HEIGHT, IMAGE_WIDTH = 224, 224
SEQUENCE_LENGTH = 16
BATCH_SIZE = 4
EPOCHS = 25
LR_BACKBONE = 1e-5
LR_HEAD = 5e-5
WEIGHT_DECAY = 0.05
NUM_FOLDS = 5
NUM_WORKERS = 4
PATIENCE = 4
PIN_MEMORY = True
USE_AMP = True
TRAIN_LAST_PERCENT = 0.50  # Best setting

DATA_DIR = "/content/drive/MyDrive/DatasetFinal"
SAVE_DIR = "hiera_modified_stages23103_last75pct"
os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 16 Classes =====================
CLASSES_LIST = [
    'Climb_Down_Stairs', 'Climb_Up_Stairs', 'Crawl', 'Drink', 'Eat',
    'Emesis', 'Exit_Crib', 'Fall_Down', 'Hold_Knife', 'Play_Slide',
    'Play_Stacking_Rings', 'Play_Swing', 'Play_Walker', 'Sleep', 'W_Sit', 'Walk'
]
NUM_CLASSES = len(CLASSES_LIST)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 70)
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
print(f"Strategy: Reduced Hiera (2,3,10,3) + Partial Pretrained + Last {TRAIN_LAST_PERCENT*100:.0f}% trainable")
print("=" * 70)


# ============================================================
# VIDEO AUGMENTATION
# ============================================================
class VideoAugmentation:
    def __init__(self, size=224, is_train=True):
        self.size = size
        self.is_train = is_train
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

    def __call__(self, frames):
        pil_frames = [transforms.ToPILImage()(f) for f in frames]

        if self.is_train:
            if random.random() < 0.5:
                pil_frames = [TF.hflip(f) for f in pil_frames]

            brightness = random.uniform(0.8, 1.2)
            contrast   = random.uniform(0.8, 1.2)
            saturation = random.uniform(0.8, 1.2)
            hue        = random.uniform(-0.05, 0.05)

            pil_frames = [TF.adjust_brightness(f, brightness) for f in pil_frames]
            pil_frames = [TF.adjust_contrast(f, contrast) for f in pil_frames]
            pil_frames = [TF.adjust_saturation(f, saturation) for f in pil_frames]
            pil_frames = [TF.adjust_hue(f, hue) for f in pil_frames]

        tensors = []
        for f in pil_frames:
            f = TF.resize(f, [self.size, self.size])
            t = TF.to_tensor(f)
            if self.is_train and random.random() < 0.3:
                noise = torch.randn_like(t) * 0.02
                t = torch.clamp(t + noise, 0.0, 1.0)
            t = self.normalize(t)
            tensors.append(t)

        return torch.stack(tensors)


def load_video_frames_raw(video_path, num_frames=16):
    container = av.open(video_path)
    frames = []
    for frame in container.decode(video=0):
        img = frame.to_ndarray(format="rgb24")
        frames.append(img)
    container.close()

    if len(frames) == 0:
        return None

    indices = np.linspace(0, len(frames) - 1, num_frames).astype(int)
    return [frames[i] for i in indices]


# ============================================================
# DATASET
# ============================================================
class ActivityDataset(Dataset):
    def __init__(self, data_dir, classes_list, num_frames=16, size=224, is_train=True):
        self.samples = []
        self.labels = []
        self.num_frames = num_frames
        self.size = size
        self.is_train = is_train
        self.class_to_idx = {cls: idx for idx, cls in enumerate(classes_list)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}
        self.augment = VideoAugmentation(size=size, is_train=is_train)

        for class_name in classes_list:
            class_path = Path(data_dir) / class_name
            if not class_path.is_dir():
                print(f"Warning: folder not found → {class_path}")
                continue
            count = 0
            for video_file in sorted(class_path.glob("*")):
                if video_file.suffix.lower() in {".mp4", ".avi", ".mov", ".mkv"}:
                    self.samples.append(str(video_file))
                    self.labels.append(self.class_to_idx[class_name])
                    count += 1
            print(f"Loaded {count} videos from '{class_name}'")

        self.labels = np.array(self.labels)
        print(f"\nTotal videos: {len(self.samples)}")
        print("Class distribution:", dict(Counter(self.labels)))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path = self.samples[idx]
        label = self.labels[idx]
        frames = load_video_frames_raw(path, self.num_frames)

        if frames is None:
            frames = [np.zeros((self.size, self.size, 3), dtype=np.uint8) for _ in range(self.num_frames)]

        video_tensor = self.augment(frames)
        return video_tensor, label, path


def collate_fn(batch):
    videos, labels, paths = zip(*batch)
    videos = torch.stack(videos)
    labels = torch.tensor(labels, dtype=torch.long)
    return videos, labels, list(paths)


# ============================================================
# FREEZE STRATEGY (Last 50%)
# ============================================================
def freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT):
    for param in model.parameters():
        param.requires_grad = False

    for param in model.head.parameters():
        param.requires_grad = True

    if hasattr(model, "blocks"):
        blocks = model.blocks
        num_blocks = len(blocks)
        num_to_train = max(1, int(num_blocks * train_last_percent))
        start_idx = num_blocks - num_to_train

        print(f"\n[Freeze Strategy] Total blocks: {num_blocks}")
        print(f"[Freeze Strategy] Training last {num_to_train} blocks "
              f"({train_last_percent*100:.0f}%) → blocks[{start_idx}:]")

        for i in range(start_idx, num_blocks):
            for param in blocks[i].parameters():
                param.requires_grad = True
    else:
        for param in model.parameters():
            param.requires_grad = True

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"[Freeze Strategy] Trainable params: {trainable:,} / {total:,} "
          f"({100 * trainable / total:.1f}%)")


# ============================================================
# PARTIAL WEIGHT LOADING + REDUCED MODEL
# ============================================================
def create_reduced_hiera_with_partial_pretrained(
    num_classes=16,
    stages=(2, 3, 10, 3),
    embed_dim=96,
    device="cuda"
):
    model = hiera.Hiera(
        input_size=(16, 224, 224),
        in_chans=3,
        embed_dim=embed_dim,
        num_heads=1,
        num_classes=num_classes,
        stages=stages,
        q_pool=3,
        q_stride=(1, 2, 2),
        mask_unit_size=(1, 8, 8),
        patch_kernel=(3, 7, 7),
        patch_stride=(2, 4, 4),
        patch_padding=(1, 3, 3),
        sep_pos_embed=True,
        drop_path_rate=0.15,
    )

    print("\nLoading original pretrained Hiera-Base weights...")
    pretrained = hiera.hiera_base_16x224(
        pretrained=True,
        checkpoint="mae_k400_ft_k400",
        num_classes=400
    )
    pretrained_dict = pretrained.state_dict()
    model_dict = model.state_dict()

    new_state_dict = OrderedDict()
    loaded_keys = []
    skipped_keys = []

    for key, value in pretrained_dict.items():
        if "head" in key:
            skipped_keys.append(key)
            continue

        if "blocks" in key:
            parts = key.split(".")
            block_idx = int(parts[1])

            if block_idx <= 4:
                new_block_idx = block_idx
            elif 5 <= block_idx <= 14:
                new_block_idx = block_idx
            elif block_idx >= 21:
                new_block_idx = block_idx - 6
            else:
                skipped_keys.append(key)
                continue

            new_key = key.replace(f"blocks.{block_idx}.", f"blocks.{new_block_idx}.")

            if new_key in model_dict and model_dict[new_key].shape == value.shape:
                new_state_dict[new_key] = value
                loaded_keys.append(new_key)
            else:
                skipped_keys.append(key)
        else:
            if key in model_dict and model_dict[key].shape == value.shape:
                new_state_dict[key] = value
                loaded_keys.append(key)
            else:
                skipped_keys.append(key)

    missing, unexpected = model.load_state_dict(new_state_dict, strict=False)

    print("\n" + "="*65)
    print("PARTIAL WEIGHT LOADING SUMMARY")
    print("="*65)
    print(f"Successfully loaded keys : {len(loaded_keys)}")
    print(f"Skipped keys             : {len(skipped_keys)}")
    print(f"Missing keys             : {len(missing)}")
    print(f"Unexpected keys          : {len(unexpected)}")
    print("="*65)

    return model.to(device)


def create_hiera_model():
    model = create_reduced_hiera_with_partial_pretrained(
        num_classes=NUM_CLASSES,
        stages=(2, 3, 10, 3),
        embed_dim=96,
        device=DEVICE
    )
    freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT)
    return model


# ============================================================
# EDGE DEPLOYMENT METRICS
# ============================================================
def print_edge_deployment_metrics(model, device=DEVICE, input_shape=(1, 3, 16, 224, 224)):
    print("\n" + "=" * 70)
    print("EDGE DEPLOYMENT METRICS")
    print("=" * 70)

    model.eval()
    model.to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    param_size_mb = total_params * 4 / (1024 ** 2)
    param_size_mb_fp16 = total_params * 2 / (1024 ** 2)

    print(f"\n[1] MEMORY FOOTPRINT")
    print(f"    Total parameters      : {total_params:,}")
    print(f"    Trainable parameters  : {trainable_params:,}")
    print(f"    Model size (FP32)     : {param_size_mb:.2f} MB")
    print(f"    Model size (FP16 est.): {param_size_mb_fp16:.2f} MB")

    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
        torch.cuda.empty_cache()
        dummy = torch.randn(*input_shape, device=device)
        with torch.no_grad():
            _ = model(dummy)
        peak_mem_mb = torch.cuda.max_memory_allocated() / (1024 ** 2)
        print(f"    Peak GPU memory (fwd) : {peak_mem_mb:.2f} MB")

    print(f"\n[2] COMPUTATIONAL COMPLEXITY (FLOPs)")
    try:
        dummy = torch.randn(*input_shape, device=device)
        flops, params = profile(model, inputs=(dummy,), verbose=False)
        flops_str, params_str = clever_format([flops, params], "%.3f")
        print(f"    FLOPs                 : {flops_str}")
        print(f"    Params (thop)         : {params_str}")
    except Exception as e:
        print(f"    FLOPs calculation failed: {e}")

    print(f"\n[3] INFERENCE TIME (Latency)")
    dummy = torch.randn(*input_shape, device=device)
    with torch.no_grad():
        for _ in range(10):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()

    n_runs = 50
    start = time.perf_counter()
    with torch.no_grad():
        for _ in range(n_runs):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    end = time.perf_counter()

    avg_latency_ms = (end - start) / n_runs * 1000
    fps = 1000.0 / avg_latency_ms
    print(f"    Avg latency (batch=1) : {avg_latency_ms:.2f} ms")
    print(f"    Throughput            : {fps:.1f} FPS")

    print(f"\n[4] POWER DRAW")
    print(f"    Measure on target edge device (Jetson / mobile NPU)")

    print(f"\n[5] THERMAL LIMITS")
    print(f"    Monitor on target hardware under sustained load")

    print(f"\n[6] MODEL ACCURACY")
    print(f"    See FINAL RESULTS section after cross-validation")
    print("=" * 70 + "\n")


# ============================================================
# LOAD FULL DATASET
# ============================================================
print("\nLoading full dataset...")
full_dataset_for_split = ActivityDataset(
    DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
)


# ============================================================
# TRAIN / EVAL
# ============================================================
def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for videos, labels, _ in tqdm(loader, desc=" train", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * videos.size(0)
        preds = logits.argmax(1)
        correct += (preds == labels).sum().item()
        total += videos.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    all_preds, all_labels, all_probs, all_paths = [], [], [], []

    for videos, labels, paths in tqdm(loader, desc=" val", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        total_loss += loss.item() * videos.size(0)
        probs = torch.softmax(logits.float(), dim=1)
        preds = logits.argmax(1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
        all_paths.extend(paths)

    return (total_loss / len(loader.dataset),
            np.array(all_preds),
            np.array(all_labels),
            np.array(all_probs),
            all_paths)


def plot_cm(cm, title, normalize=True):
    if normalize:
        cm_plot = cm.astype("float") / (cm.sum(axis=1)[:, np.newaxis] + 1e-8)
        fmt = ".2f"
    else:
        cm_plot = cm
        fmt = "d"

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm_plot, annot=True, fmt=fmt, cmap="Blues",
                xticklabels=CLASSES_LIST, yticklabels=CLASSES_LIST)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# PROFILE MODEL ONCE
# ============================================================
print("\nCreating temporary model for edge metrics...")
_temp_model = create_hiera_model()
print_edge_deployment_metrics(_temp_model)
del _temp_model
torch.cuda.empty_cache()


# ============================================================
# 5-FOLD CROSS VALIDATION
# ============================================================
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)

accuracy_scores = []
top2_scores = []
f1_macro_scores = []
f1_weighted_scores = []
precision_scores = []
recall_scores = []
auc_scores = []
confusion_matrices = []

overall_best_f1 = 0.0
overall_best_path = None

fold = 1
for train_idx, val_idx in skf.split(np.zeros(len(full_dataset_for_split)), full_dataset_for_split.labels):
    print(f"\n{'='*70}")
    print(f"========== Fold {fold}/{NUM_FOLDS} ==========")
    print(f"{'='*70}")

    train_dataset = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=True)
    val_dataset   = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False)

    train_subset = Subset(train_dataset, train_idx)
    val_subset   = Subset(val_dataset, val_idx)

    train_labels = full_dataset_for_split.labels[train_idx]
    class_weights = compute_class_weight("balanced", classes=np.unique(train_labels), y=train_labels)
    class_weight_tensor = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
    print("Class weights (this fold):", dict(enumerate(np.round(class_weights, 3))))

    train_loader = DataLoader(
        train_subset, batch_size=BATCH_SIZE, shuffle=True,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )
    val_loader = DataLoader(
        val_subset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )

    model = create_hiera_model()
    criterion = nn.CrossEntropyLoss(weight=class_weight_tensor)

    backbone_params = [p for n, p in model.named_parameters() if "head" not in n and p.requires_grad]
    head_params     = [p for n, p in model.named_parameters() if "head" in n and p.requires_grad]

    optimizer = torch.optim.AdamW([
        {"params": backbone_params, "lr": LR_BACKBONE},
        {"params": head_params,     "lr": LR_HEAD}
    ], weight_decay=WEIGHT_DECAY)

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    best_f1 = 0.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        val_loss, y_pred, y_true, y_prob, _ = evaluate(model, val_loader, criterion)
        scheduler.step()

        current_lrs = [group['lr'] for group in optimizer.param_groups]
        print(f"  → LRs: backbone={current_lrs[0]:.2e} | head={current_lrs[1]:.2e}")

        val_acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average="macro", zero_division=0)

        try:
            top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
        except Exception:
            top2_acc = 0.0

        print(f"Epoch {epoch+1:02d} | "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} "
              f"top2={top2_acc:.4f} macro_F1={f1_macro:.4f}")

        if f1_macro > best_f1:
            best_f1 = f1_macro
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
            print(f" → New best macro F1: {best_f1:.4f}")
        else:
            epochs_no_improve += 1
            print(f" → No improvement for {epochs_no_improve}/{PATIENCE} epochs")

        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch+1}")
            break

    model.load_state_dict(best_state)
    model.to(DEVICE)
    val_loss, y_pred, y_true, y_prob, video_paths = evaluate(model, val_loader, criterion)

    acc = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall_macro    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_macro        = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_weighted     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    try:
        top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
    except Exception:
        top2_acc = 0.0

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc_macro = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except Exception as e:
        print(f"AUC calculation warning: {e}")
        auc_macro = 0.0

    accuracy_scores.append(acc)
    top2_scores.append(top2_acc)
    f1_macro_scores.append(f1_macro)
    f1_weighted_scores.append(f1_weighted)
    precision_scores.append(precision_macro)
    recall_scores.append(recall_macro)
    auc_scores.append(auc_macro)

    cm = confusion_matrix(y_true, y_pred)
    confusion_matrices.append(cm)
    plot_cm(cm, f"Activity Recognition CM – Fold {fold}")

    print(f"\n{'='*60}")
    print(f"Fold {fold} Final Metrics:")
    print(f"{'='*60}")
    print(f"Accuracy        : {acc:.4f}")
    print(f"Top-2 Accuracy  : {top2_acc:.4f}")
    print(f"Precision (Macro): {precision_macro:.4f}")
    print(f"Recall (Macro)  : {recall_macro:.4f}")
    print(f"F1-Score (Macro): {f1_macro:.4f}")
    print(f"F1-Score (Weighted): {f1_weighted:.4f}")
    print(f"AUC (Macro OVR) : {auc_macro:.4f}")
    print(f"{'='*60}")

    print(f"\nFold {fold} Classification Report:")
    print(classification_report(y_true, y_pred, target_names=CLASSES_LIST, digits=4, zero_division=0))

    true_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_true]
    pred_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_pred]

    df = pd.DataFrame({
        "Video_Path": video_paths,
        "True_Class": true_class_names,
        "Predicted_Class": pred_class_names,
        "Correct": [t == p for t, p in zip(true_class_names, pred_class_names)]
    })

    excel_path = os.path.join(SAVE_DIR, f"predictions_fold_{fold}.xlsx")
    df.to_excel(excel_path, index=False)
    print(f"Excel saved → {excel_path}")

    fold_save_path = os.path.join(SAVE_DIR, f"best_hiera_modified_fold_{fold}.pth")
    torch.save({
        "model_state_dict": best_state,
        "fold": fold,
        "macro_f1": best_f1,
        "accuracy": acc,
        "top2_accuracy": top2_acc,
        "class_to_idx": full_dataset_for_split.class_to_idx,
        "idx_to_class": full_dataset_for_split.idx_to_class,
        "classes": CLASSES_LIST,
        "stages": (2, 3, 10, 3),
    }, fold_save_path)
    print(f"Model saved → {fold_save_path}")

    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1
        overall_best_path = fold_save_path

    del optimizer, scaler, model
    torch.cuda.empty_cache()
    fold += 1


# ============================================================
# FINAL RESULTS
# ============================================================
print("\n" + "="*70)
print("FINAL RESULTS (Modified Hiera - stages=(2,3,10,3) + Partial Pretrained + Last 75%)")
print("="*70)

print(f"\nPer-fold Accuracy : {[f'{x:.4f}' for x in accuracy_scores]}")
print(f"Mean ± std Accuracy : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")

print(f"\nPer-fold Top-2 Accuracy : {[f'{x:.4f}' for x in top2_scores]}")
print(f"Mean ± std Top-2 : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")

print(f"\nPer-fold Precision : {[f'{x:.4f}' for x in precision_scores]}")
print(f"Mean ± std Precision : {np.mean(precision_scores):.4f} ± {np.std(precision_scores):.4f}")

print(f"\nPer-fold Recall : {[f'{x:.4f}' for x in recall_scores]}")
print(f"Mean ± std Recall : {np.mean(recall_scores):.4f} ± {np.std(recall_scores):.4f}")

print(f"\nPer-fold Macro F1 : {[f'{x:.4f}' for x in f1_macro_scores]}")
print(f"Mean ± std Macro F1 : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")

print(f"\nPer-fold Weighted F1 : {[f'{x:.4f}' for x in f1_weighted_scores]}")
print(f"Mean ± std Weighted F1 : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")

print(f"\nPer-fold AUC (Macro) : {[f'{x:.4f}' for x in auc_scores]}")
print(f"Mean ± std AUC : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")

conf_sum = np.sum(confusion_matrices, axis=0)
plot_cm(conf_sum, "Aggregated Confusion Matrix (All Folds)", normalize=False)
plot_cm(conf_sum, "Aggregated Confusion Matrix (Normalized)", normalize=True)

print(f"\nOverall best model (Macro F1={overall_best_f1:.4f}) saved at:")
print(overall_best_path)

print("\n" + "=" * 70)
print("EDGE DEPLOYMENT SUMMARY (with measured accuracy)")
print("=" * 70)
print(f"Model Architecture     : Modified Hiera stages=(2,3,10,3)")
print(f"Pretrained Loading     : Partial (from original Hiera-Base)")
print(f"Trainable Strategy     : Last 50% blocks + Head")
print(f"Mean Accuracy          : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")
print(f"Mean Top-2 Accuracy    : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")
print(f"Mean Macro F1          : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")
print(f"Mean Weighted F1       : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")
print(f"Mean AUC               : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("=" * 70)
```

```python
# ============================================================
# Toddler Activity Recognition - Modified Hiera
# Reduced stages=(2,3,5,3) + Partial Pretrained Loading + Last 50% Trainable
# + Class weights inside each fold
# + Consistent Video Augmentations
# + LR logging + Top-2 Accuracy + Edge Deployment Metrics
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

!pip install -q torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
!pip install -q hiera-transformer av opencv-python tqdm scikit-learn seaborn pandas openpyxl thop

import os
import random
import time
from pathlib import Path
from collections import Counter, OrderedDict

import av
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.transforms import functional as TF
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    precision_score, recall_score, roc_auc_score, top_k_accuracy_score
)
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns
import hiera
from thop import profile, clever_format

# ────────────────────────────────────────────────
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    torch.backends.cudnn.benchmark = True

# ===================== Settings =====================
IMAGE_HEIGHT, IMAGE_WIDTH = 224, 224
SEQUENCE_LENGTH = 16
BATCH_SIZE = 4
EPOCHS = 25
LR_BACKBONE = 1e-5
LR_HEAD = 5e-5
WEIGHT_DECAY = 0.05
NUM_FOLDS = 5
NUM_WORKERS = 4
PATIENCE = 4
PIN_MEMORY = True
USE_AMP = True
TRAIN_LAST_PERCENT = 0.50  # Best setting

DATA_DIR = "/content/drive/MyDrive/DatasetFinal"
SAVE_DIR = "hiera_modified_stages2353_last50pct"
os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 16 Classes =====================
CLASSES_LIST = [
    'Climb_Down_Stairs', 'Climb_Up_Stairs', 'Crawl', 'Drink', 'Eat',
    'Emesis', 'Exit_Crib', 'Fall_Down', 'Hold_Knife', 'Play_Slide',
    'Play_Stacking_Rings', 'Play_Swing', 'Play_Walker', 'Sleep', 'W_Sit', 'Walk'
]
NUM_CLASSES = len(CLASSES_LIST)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 70)
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
print(f"Strategy: Reduced Hiera (2,3,5,3) + Partial Pretrained + Last {TRAIN_LAST_PERCENT*100:.0f}% trainable")
print("=" * 70)


# ============================================================
# VIDEO AUGMENTATION
# ============================================================
class VideoAugmentation:
    def __init__(self, size=224, is_train=True):
        self.size = size
        self.is_train = is_train
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

    def __call__(self, frames):
        pil_frames = [transforms.ToPILImage()(f) for f in frames]

        if self.is_train:
            if random.random() < 0.5:
                pil_frames = [TF.hflip(f) for f in pil_frames]

            brightness = random.uniform(0.8, 1.2)
            contrast   = random.uniform(0.8, 1.2)
            saturation = random.uniform(0.8, 1.2)
            hue        = random.uniform(-0.05, 0.05)

            pil_frames = [TF.adjust_brightness(f, brightness) for f in pil_frames]
            pil_frames = [TF.adjust_contrast(f, contrast) for f in pil_frames]
            pil_frames = [TF.adjust_saturation(f, saturation) for f in pil_frames]
            pil_frames = [TF.adjust_hue(f, hue) for f in pil_frames]

        tensors = []
        for f in pil_frames:
            f = TF.resize(f, [self.size, self.size])
            t = TF.to_tensor(f)
            if self.is_train and random.random() < 0.3:
                noise = torch.randn_like(t) * 0.02
                t = torch.clamp(t + noise, 0.0, 1.0)
            t = self.normalize(t)
            tensors.append(t)

        return torch.stack(tensors)


def load_video_frames_raw(video_path, num_frames=16):
    container = av.open(video_path)
    frames = []
    for frame in container.decode(video=0):
        img = frame.to_ndarray(format="rgb24")
        frames.append(img)
    container.close()

    if len(frames) == 0:
        return None

    indices = np.linspace(0, len(frames) - 1, num_frames).astype(int)
    return [frames[i] for i in indices]


# ============================================================
# DATASET
# ============================================================
class ActivityDataset(Dataset):
    def __init__(self, data_dir, classes_list, num_frames=16, size=224, is_train=True):
        self.samples = []
        self.labels = []
        self.num_frames = num_frames
        self.size = size
        self.is_train = is_train
        self.class_to_idx = {cls: idx for idx, cls in enumerate(classes_list)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}
        self.augment = VideoAugmentation(size=size, is_train=is_train)

        for class_name in classes_list:
            class_path = Path(data_dir) / class_name
            if not class_path.is_dir():
                print(f"Warning: folder not found → {class_path}")
                continue
            count = 0
            for video_file in sorted(class_path.glob("*")):
                if video_file.suffix.lower() in {".mp4", ".avi", ".mov", ".mkv"}:
                    self.samples.append(str(video_file))
                    self.labels.append(self.class_to_idx[class_name])
                    count += 1
            print(f"Loaded {count} videos from '{class_name}'")

        self.labels = np.array(self.labels)
        print(f"\nTotal videos: {len(self.samples)}")
        print("Class distribution:", dict(Counter(self.labels)))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path = self.samples[idx]
        label = self.labels[idx]
        frames = load_video_frames_raw(path, self.num_frames)

        if frames is None:
            frames = [np.zeros((self.size, self.size, 3), dtype=np.uint8) for _ in range(self.num_frames)]

        video_tensor = self.augment(frames)
        return video_tensor, label, path


def collate_fn(batch):
    videos, labels, paths = zip(*batch)
    videos = torch.stack(videos)
    labels = torch.tensor(labels, dtype=torch.long)
    return videos, labels, list(paths)


# ============================================================
# FREEZE STRATEGY (Last 50%)
# ============================================================
def freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT):
    for param in model.parameters():
        param.requires_grad = False

    for param in model.head.parameters():
        param.requires_grad = True

    if hasattr(model, "blocks"):
        blocks = model.blocks
        num_blocks = len(blocks)
        num_to_train = max(1, int(num_blocks * train_last_percent))
        start_idx = num_blocks - num_to_train

        print(f"\n[Freeze Strategy] Total blocks: {num_blocks}")
        print(f"[Freeze Strategy] Training last {num_to_train} blocks "
              f"({train_last_percent*100:.0f}%) → blocks[{start_idx}:]")

        for i in range(start_idx, num_blocks):
            for param in blocks[i].parameters():
                param.requires_grad = True
    else:
        for param in model.parameters():
            param.requires_grad = True

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"[Freeze Strategy] Trainable params: {trainable:,} / {total:,} "
          f"({100 * trainable / total:.1f}%)")


# ============================================================
# PARTIAL WEIGHT LOADING + REDUCED MODEL  (stages=(2,3,5,3))
# ============================================================
def create_reduced_hiera_with_partial_pretrained(
    num_classes=16,
    stages=(2, 3, 5, 3),          # ← reduced from (2,3,10,3)
    embed_dim=96,
    device="cuda"
):
    model = hiera.Hiera(
        input_size=(16, 224, 224),
        in_chans=3,
        embed_dim=embed_dim,
        num_heads=1,
        num_classes=num_classes,
        stages=stages,
        q_pool=3,
        q_stride=(1, 2, 2),
        mask_unit_size=(1, 8, 8),
        patch_kernel=(3, 7, 7),
        patch_stride=(2, 4, 4),
        patch_padding=(1, 3, 3),
        sep_pos_embed=True,
        drop_path_rate=0.15,
    )

    print("\nLoading original pretrained Hiera-Base weights...")
    pretrained = hiera.hiera_base_16x224(
        pretrained=True,
        checkpoint="mae_k400_ft_k400",
        num_classes=400
    )
    pretrained_dict = pretrained.state_dict()
    model_dict = model.state_dict()

    new_state_dict = OrderedDict()
    loaded_keys = []
    skipped_keys = []

    # Original Hiera-Base stages ≈ (2, 3, 16, 3) → 24 blocks
    # Our model stages = (2, 3, 5, 3) → 13 blocks
    # Mapping strategy (keep early & late, drop middle of stage-3):
    #   blocks 0-4   → stay 0-4          (stage 1+2)
    #   blocks 5-9   → map to 5-9        (first 5 of original stage-3)
    #   blocks 21-23 → map to 10-12      (last 3 blocks)
    for key, value in pretrained_dict.items():
        if "head" in key:
            skipped_keys.append(key)
            continue

        if "blocks" in key:
            parts = key.split(".")
            block_idx = int(parts[1])

            if block_idx <= 4:
                new_block_idx = block_idx
            elif 5 <= block_idx <= 9:
                new_block_idx = block_idx          # keep first 5 of stage-3
            elif block_idx >= 21:
                new_block_idx = block_idx - 11     # 21→10, 22→11, 23→12
            else:
                skipped_keys.append(key)
                continue

            new_key = key.replace(f"blocks.{block_idx}.", f"blocks.{new_block_idx}.")

            if new_key in model_dict and model_dict[new_key].shape == value.shape:
                new_state_dict[new_key] = value
                loaded_keys.append(new_key)
            else:
                skipped_keys.append(key)
        else:
            if key in model_dict and model_dict[key].shape == value.shape:
                new_state_dict[key] = value
                loaded_keys.append(key)
            else:
                skipped_keys.append(key)

    missing, unexpected = model.load_state_dict(new_state_dict, strict=False)

    print("\n" + "="*65)
    print("PARTIAL WEIGHT LOADING SUMMARY")
    print("="*65)
    print(f"Successfully loaded keys : {len(loaded_keys)}")
    print(f"Skipped keys             : {len(skipped_keys)}")
    print(f"Missing keys             : {len(missing)}")
    print(f"Unexpected keys          : {len(unexpected)}")
    print("="*65)

    return model.to(device)


def create_hiera_model():
    model = create_reduced_hiera_with_partial_pretrained(
        num_classes=NUM_CLASSES,
        stages=(2, 3, 5, 3),          # ← reduced
        embed_dim=96,
        device=DEVICE
    )
    freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT)
    return model


# ============================================================
# EDGE DEPLOYMENT METRICS
# ============================================================
def print_edge_deployment_metrics(model, device=DEVICE, input_shape=(1, 3, 16, 224, 224)):
    print("\n" + "=" * 70)
    print("EDGE DEPLOYMENT METRICS")
    print("=" * 70)

    model.eval()
    model.to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    param_size_mb = total_params * 4 / (1024 ** 2)
    param_size_mb_fp16 = total_params * 2 / (1024 ** 2)

    print(f"\n[1] MEMORY FOOTPRINT")
    print(f"    Total parameters      : {total_params:,}")
    print(f"    Trainable parameters  : {trainable_params:,}")
    print(f"    Model size (FP32)     : {param_size_mb:.2f} MB")
    print(f"    Model size (FP16 est.): {param_size_mb_fp16:.2f} MB")

    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
        torch.cuda.empty_cache()
        dummy = torch.randn(*input_shape, device=device)
        with torch.no_grad():
            _ = model(dummy)
        peak_mem_mb = torch.cuda.max_memory_allocated() / (1024 ** 2)
        print(f"    Peak GPU memory (fwd) : {peak_mem_mb:.2f} MB")

    print(f"\n[2] COMPUTATIONAL COMPLEXITY (FLOPs)")
    try:
        dummy = torch.randn(*input_shape, device=device)
        flops, params = profile(model, inputs=(dummy,), verbose=False)
        flops_str, params_str = clever_format([flops, params], "%.3f")
        print(f"    FLOPs                 : {flops_str}")
        print(f"    Params (thop)         : {params_str}")
    except Exception as e:
        print(f"    FLOPs calculation failed: {e}")

    print(f"\n[3] INFERENCE TIME (Latency)")
    dummy = torch.randn(*input_shape, device=device)
    with torch.no_grad():
        for _ in range(10):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()

    n_runs = 50
    start = time.perf_counter()
    with torch.no_grad():
        for _ in range(n_runs):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    end = time.perf_counter()

    avg_latency_ms = (end - start) / n_runs * 1000
    fps = 1000.0 / avg_latency_ms
    print(f"    Avg latency (batch=1) : {avg_latency_ms:.2f} ms")
    print(f"    Throughput            : {fps:.1f} FPS")

    print(f"\n[4] POWER DRAW")
    print(f"    Measure on target edge device (Jetson / mobile NPU)")

    print(f"\n[5] THERMAL LIMITS")
    print(f"    Monitor on target hardware under sustained load")

    print(f"\n[6] MODEL ACCURACY")
    print(f"    See FINAL RESULTS section after cross-validation")
    print("=" * 70 + "\n")


# ============================================================
# LOAD FULL DATASET
# ============================================================
print("\nLoading full dataset...")
full_dataset_for_split = ActivityDataset(
    DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
)


# ============================================================
# TRAIN / EVAL
# ============================================================
def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for videos, labels, _ in tqdm(loader, desc=" train", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * videos.size(0)
        preds = logits.argmax(1)
        correct += (preds == labels).sum().item()
        total += videos.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    all_preds, all_labels, all_probs, all_paths = [], [], [], []

    for videos, labels, paths in tqdm(loader, desc=" val", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        total_loss += loss.item() * videos.size(0)
        probs = torch.softmax(logits.float(), dim=1)
        preds = logits.argmax(1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
        all_paths.extend(paths)

    return (total_loss / len(loader.dataset),
            np.array(all_preds),
            np.array(all_labels),
            np.array(all_probs),
            all_paths)


def plot_cm(cm, title, normalize=True):
    if normalize:
        cm_plot = cm.astype("float") / (cm.sum(axis=1)[:, np.newaxis] + 1e-8)
        fmt = ".2f"
    else:
        cm_plot = cm
        fmt = "d"

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm_plot, annot=True, fmt=fmt, cmap="Blues",
                xticklabels=CLASSES_LIST, yticklabels=CLASSES_LIST)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# PROFILE MODEL ONCE
# ============================================================
print("\nCreating temporary model for edge metrics...")
_temp_model = create_hiera_model()
print_edge_deployment_metrics(_temp_model)
del _temp_model
torch.cuda.empty_cache()


# ============================================================
# 5-FOLD CROSS VALIDATION
# ============================================================
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)

accuracy_scores = []
top2_scores = []
f1_macro_scores = []
f1_weighted_scores = []
precision_scores = []
recall_scores = []
auc_scores = []
confusion_matrices = []

overall_best_f1 = 0.0
overall_best_path = None

fold = 1
for train_idx, val_idx in skf.split(np.zeros(len(full_dataset_for_split)), full_dataset_for_split.labels):
    print(f"\n{'='*70}")
    print(f"========== Fold {fold}/{NUM_FOLDS} ==========")
    print(f"{'='*70}")

    train_dataset = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=True)
    val_dataset   = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False)

    train_subset = Subset(train_dataset, train_idx)
    val_subset   = Subset(val_dataset, val_idx)

    train_labels = full_dataset_for_split.labels[train_idx]
    class_weights = compute_class_weight("balanced", classes=np.unique(train_labels), y=train_labels)
    class_weight_tensor = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
    print("Class weights (this fold):", dict(enumerate(np.round(class_weights, 3))))

    train_loader = DataLoader(
        train_subset, batch_size=BATCH_SIZE, shuffle=True,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )
    val_loader = DataLoader(
        val_subset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )

    model = create_hiera_model()
    criterion = nn.CrossEntropyLoss(weight=class_weight_tensor)

    backbone_params = [p for n, p in model.named_parameters() if "head" not in n and p.requires_grad]
    head_params     = [p for n, p in model.named_parameters() if "head" in n and p.requires_grad]

    optimizer = torch.optim.AdamW([
        {"params": backbone_params, "lr": LR_BACKBONE},
        {"params": head_params,     "lr": LR_HEAD}
    ], weight_decay=WEIGHT_DECAY)

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    best_f1 = 0.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        val_loss, y_pred, y_true, y_prob, _ = evaluate(model, val_loader, criterion)
        scheduler.step()

        current_lrs = [group['lr'] for group in optimizer.param_groups]
        print(f"  → LRs: backbone={current_lrs[0]:.2e} | head={current_lrs[1]:.2e}")

        val_acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average="macro", zero_division=0)

        try:
            top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
        except Exception:
            top2_acc = 0.0

        print(f"Epoch {epoch+1:02d} | "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} "
              f"top2={top2_acc:.4f} macro_F1={f1_macro:.4f}")

        if f1_macro > best_f1:
            best_f1 = f1_macro
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
            print(f" → New best macro F1: {best_f1:.4f}")
        else:
            epochs_no_improve += 1
            print(f" → No improvement for {epochs_no_improve}/{PATIENCE} epochs")

        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch+1}")
            break

    model.load_state_dict(best_state)
    model.to(DEVICE)
    val_loss, y_pred, y_true, y_prob, video_paths = evaluate(model, val_loader, criterion)

    acc = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall_macro    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_macro        = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_weighted     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    try:
        top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
    except Exception:
        top2_acc = 0.0

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc_macro = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except Exception as e:
        print(f"AUC calculation warning: {e}")
        auc_macro = 0.0

    accuracy_scores.append(acc)
    top2_scores.append(top2_acc)
    f1_macro_scores.append(f1_macro)
    f1_weighted_scores.append(f1_weighted)
    precision_scores.append(precision_macro)
    recall_scores.append(recall_macro)
    auc_scores.append(auc_macro)

    cm = confusion_matrix(y_true, y_pred)
    confusion_matrices.append(cm)
    plot_cm(cm, f"Activity Recognition CM – Fold {fold}")

    print(f"\n{'='*60}")
    print(f"Fold {fold} Final Metrics:")
    print(f"{'='*60}")
    print(f"Accuracy        : {acc:.4f}")
    print(f"Top-2 Accuracy  : {top2_acc:.4f}")
    print(f"Precision (Macro): {precision_macro:.4f}")
    print(f"Recall (Macro)  : {recall_macro:.4f}")
    print(f"F1-Score (Macro): {f1_macro:.4f}")
    print(f"F1-Score (Weighted): {f1_weighted:.4f}")
    print(f"AUC (Macro OVR) : {auc_macro:.4f}")
    print(f"{'='*60}")

    print(f"\nFold {fold} Classification Report:")
    print(classification_report(y_true, y_pred, target_names=CLASSES_LIST, digits=4, zero_division=0))

    true_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_true]
    pred_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_pred]

    df = pd.DataFrame({
        "Video_Path": video_paths,
        "True_Class": true_class_names,
        "Predicted_Class": pred_class_names,
        "Correct": [t == p for t, p in zip(true_class_names, pred_class_names)]
    })

    excel_path = os.path.join(SAVE_DIR, f"predictions_fold_{fold}.xlsx")
    df.to_excel(excel_path, index=False)
    print(f"Excel saved → {excel_path}")

    fold_save_path = os.path.join(SAVE_DIR, f"best_hiera_modified_fold_{fold}.pth")
    torch.save({
        "model_state_dict": best_state,
        "fold": fold,
        "macro_f1": best_f1,
        "accuracy": acc,
        "top2_accuracy": top2_acc,
        "class_to_idx": full_dataset_for_split.class_to_idx,
        "idx_to_class": full_dataset_for_split.idx_to_class,
        "classes": CLASSES_LIST,
        "stages": (2, 3, 5, 3),          # ← updated
    }, fold_save_path)
    print(f"Model saved → {fold_save_path}")

    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1
        overall_best_path = fold_save_path

    del optimizer, scaler, model
    torch.cuda.empty_cache()
    fold += 1


# ============================================================
# FINAL RESULTS
# ============================================================
print("\n" + "="*70)
print("FINAL RESULTS (Modified Hiera - stages=(2,3,5,3) + Partial Pretrained + Last 50%)")
print("="*70)

print(f"\nPer-fold Accuracy : {[f'{x:.4f}' for x in accuracy_scores]}")
print(f"Mean ± std Accuracy : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")

print(f"\nPer-fold Top-2 Accuracy : {[f'{x:.4f}' for x in top2_scores]}")
print(f"Mean ± std Top-2 : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")

print(f"\nPer-fold Precision : {[f'{x:.4f}' for x in precision_scores]}")
print(f"Mean ± std Precision : {np.mean(precision_scores):.4f} ± {np.std(precision_scores):.4f}")

print(f"\nPer-fold Recall : {[f'{x:.4f}' for x in recall_scores]}")
print(f"Mean ± std Recall : {np.mean(recall_scores):.4f} ± {np.std(recall_scores):.4f}")

print(f"\nPer-fold Macro F1 : {[f'{x:.4f}' for x in f1_macro_scores]}")
print(f"Mean ± std Macro F1 : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")

print(f"\nPer-fold Weighted F1 : {[f'{x:.4f}' for x in f1_weighted_scores]}")
print(f"Mean ± std Weighted F1 : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")

print(f"\nPer-fold AUC (Macro) : {[f'{x:.4f}' for x in auc_scores]}")
print(f"Mean ± std AUC : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")

conf_sum = np.sum(confusion_matrices, axis=0)
plot_cm(conf_sum, "Aggregated Confusion Matrix (All Folds)", normalize=False)
plot_cm(conf_sum, "Aggregated Confusion Matrix (Normalized)", normalize=True)

print(f"\nOverall best model (Macro F1={overall_best_f1:.4f}) saved at:")
print(overall_best_path)

print("\n" + "=" * 70)
print("EDGE DEPLOYMENT SUMMARY (with measured accuracy)")
print("=" * 70)
print(f"Model Architecture     : Modified Hiera stages=(2,3,5,3)")
print(f"Pretrained Loading     : Partial (from original Hiera-Base)")
print(f"Trainable Strategy     : Last 50% blocks + Head")
print(f"Mean Accuracy          : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")
print(f"Mean Top-2 Accuracy    : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")
print(f"Mean Macro F1          : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")
print(f"Mean Weighted F1       : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")
print(f"Mean AUC               : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("=" * 70)
```

```python
# ============================================================
# Toddler Activity Recognition - Modified Hiera
# batch size 4
# Reduced stages=(2,3,11,3) + Partial Pretrained Loading + FULL TRAINING
# + Class weights inside each fold
# + Consistent Video Augmentations
# + LR logging + Top-2 Accuracy + Edge Deployment Metrics
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

!pip install -q torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
!pip install -q hiera-transformer av opencv-python tqdm scikit-learn seaborn pandas openpyxl thop

import os
import random
import time
from pathlib import Path
from collections import Counter, OrderedDict

import av
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.transforms import functional as TF
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    precision_score, recall_score, roc_auc_score, top_k_accuracy_score
)
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns
import hiera
from thop import profile, clever_format

# ────────────────────────────────────────────────
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    torch.backends.cudnn.benchmark = True

# ===================== Settings =====================
IMAGE_HEIGHT, IMAGE_WIDTH = 224, 224
SEQUENCE_LENGTH = 16
BATCH_SIZE = 4
EPOCHS = 25
LR_BACKBONE = 1e-5
LR_HEAD = 5e-5
WEIGHT_DECAY = 0.05
NUM_FOLDS = 5
NUM_WORKERS = 4
PATIENCE = 4
PIN_MEMORY = True
USE_AMP = True
TRAIN_LAST_PERCENT = 1.0          # FULL TRAINING (all blocks + head)

DATA_DIR = "/content/drive/MyDrive/DatasetFinal"
SAVE_DIR = "hiera_modified_stages23113_full_train"
os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 16 Classes =====================
CLASSES_LIST = [
    'Climb_Down_Stairs', 'Climb_Up_Stairs', 'Crawl', 'Drink', 'Eat',
    'Emesis', 'Exit_Crib', 'Fall_Down', 'Hold_Knife', 'Play_Slide',
    'Play_Stacking_Rings', 'Play_Swing', 'Play_Walker', 'Sleep', 'W_Sit', 'Walk'
]
NUM_CLASSES = len(CLASSES_LIST)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 70)
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
print(f"Strategy: Reduced Hiera (2,3,11,3) + Partial Pretrained + FULL TRAINING")
print("=" * 70)


# ============================================================
# VIDEO AUGMENTATION
# ============================================================
class VideoAugmentation:
    def __init__(self, size=224, is_train=True):
        self.size = size
        self.is_train = is_train
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

    def __call__(self, frames):
        pil_frames = [transforms.ToPILImage()(f) for f in frames]

        if self.is_train:
            if random.random() < 0.5:
                pil_frames = [TF.hflip(f) for f in pil_frames]

            brightness = random.uniform(0.8, 1.2)
            contrast   = random.uniform(0.8, 1.2)
            saturation = random.uniform(0.8, 1.2)
            hue        = random.uniform(-0.05, 0.05)

            pil_frames = [TF.adjust_brightness(f, brightness) for f in pil_frames]
            pil_frames = [TF.adjust_contrast(f, contrast) for f in pil_frames]
            pil_frames = [TF.adjust_saturation(f, saturation) for f in pil_frames]
            pil_frames = [TF.adjust_hue(f, hue) for f in pil_frames]

        tensors = []
        for f in pil_frames:
            f = TF.resize(f, [self.size, self.size])
            t = TF.to_tensor(f)
            if self.is_train and random.random() < 0.3:
                noise = torch.randn_like(t) * 0.02
                t = torch.clamp(t + noise, 0.0, 1.0)
            t = self.normalize(t)
            tensors.append(t)

        return torch.stack(tensors)


def load_video_frames_raw(video_path, num_frames=16):
    container = av.open(video_path)
    frames = []
    for frame in container.decode(video=0):
        img = frame.to_ndarray(format="rgb24")
        frames.append(img)
    container.close()

    if len(frames) == 0:
        return None

    indices = np.linspace(0, len(frames) - 1, num_frames).astype(int)
    return [frames[i] for i in indices]


# ============================================================
# DATASET
# ============================================================
class ActivityDataset(Dataset):
    def __init__(self, data_dir, classes_list, num_frames=16, size=224, is_train=True):
        self.samples = []
        self.labels = []
        self.num_frames = num_frames
        self.size = size
        self.is_train = is_train
        self.class_to_idx = {cls: idx for idx, cls in enumerate(classes_list)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}
        self.augment = VideoAugmentation(size=size, is_train=is_train)

        for class_name in classes_list:
            class_path = Path(data_dir) / class_name
            if not class_path.is_dir():
                print(f"Warning: folder not found → {class_path}")
                continue
            count = 0
            for video_file in sorted(class_path.glob("*")):
                if video_file.suffix.lower() in {".mp4", ".avi", ".mov", ".mkv"}:
                    self.samples.append(str(video_file))
                    self.labels.append(self.class_to_idx[class_name])
                    count += 1
            print(f"Loaded {count} videos from '{class_name}'")

        self.labels = np.array(self.labels)
        print(f"\nTotal videos: {len(self.samples)}")
        print("Class distribution:", dict(Counter(self.labels)))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path = self.samples[idx]
        label = self.labels[idx]
        frames = load_video_frames_raw(path, self.num_frames)

        if frames is None:
            frames = [np.zeros((self.size, self.size, 3), dtype=np.uint8) for _ in range(self.num_frames)]

        video_tensor = self.augment(frames)
        return video_tensor, label, path


def collate_fn(batch):
    videos, labels, paths = zip(*batch)
    videos = torch.stack(videos)
    labels = torch.tensor(labels, dtype=torch.long)
    return videos, labels, list(paths)


# ============================================================
# FULL TRAINING (no freezing)
# ============================================================
def freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT):
    """
    Full training mode: every parameter is trainable.
    The train_last_percent argument is kept only for API compatibility.
    """
    for param in model.parameters():
        param.requires_grad = True

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"\n[Full Training] Trainable params: {trainable:,} / {total:,} "
          f"({100 * trainable / total:.1f}%)")


# ============================================================
# PARTIAL WEIGHT LOADING + REDUCED MODEL
# ============================================================
def create_reduced_hiera_with_partial_pretrained(
    num_classes=16,
    stages=(2, 3, 11, 3),
    embed_dim=96,
    device="cuda"
):
    model = hiera.Hiera(
        input_size=(16, 224, 224),
        in_chans=3,
        embed_dim=embed_dim,
        num_heads=1,
        num_classes=num_classes,
        stages=stages,
        q_pool=3,
        q_stride=(1, 2, 2),
        mask_unit_size=(1, 8, 8),
        patch_kernel=(3, 7, 7),
        patch_stride=(2, 4, 4),
        patch_padding=(1, 3, 3),
        sep_pos_embed=True,
        drop_path_rate=0.15,
    )

    print("\nLoading original pretrained Hiera-Base weights...")
    pretrained = hiera.hiera_base_16x224(
        pretrained=True,
        checkpoint="mae_k400_ft_k400",
        num_classes=400
    )
    pretrained_dict = pretrained.state_dict()
    model_dict = model.state_dict()

    new_state_dict = OrderedDict()
    loaded_keys = []
    skipped_keys = []

    # Original Hiera-Base stages ≈ (2, 3, 16, 3) → 24 blocks
    # Target stages = (2, 3, 11, 3) → 19 blocks
    # Mapping:
    #   blocks 0-4   (stage1+2)  → keep
    #   blocks 5-15  (11 of stage3) → keep
    #   blocks 16-20 (remaining stage3) → skip
    #   blocks 21-23 (stage4) → shift by -5 → 16-18
    for key, value in pretrained_dict.items():
        if "head" in key:
            skipped_keys.append(key)
            continue

        if "blocks" in key:
            parts = key.split(".")
            block_idx = int(parts[1])

            if block_idx <= 4:
                new_block_idx = block_idx
            elif 5 <= block_idx <= 15:
                new_block_idx = block_idx
            elif block_idx >= 21:
                new_block_idx = block_idx - 5
            else:
                skipped_keys.append(key)
                continue

            new_key = key.replace(f"blocks.{block_idx}.", f"blocks.{new_block_idx}.")

            if new_key in model_dict and model_dict[new_key].shape == value.shape:
                new_state_dict[new_key] = value
                loaded_keys.append(new_key)
            else:
                skipped_keys.append(key)
        else:
            if key in model_dict and model_dict[key].shape == value.shape:
                new_state_dict[key] = value
                loaded_keys.append(key)
            else:
                skipped_keys.append(key)

    missing, unexpected = model.load_state_dict(new_state_dict, strict=False)

    print("\n" + "="*65)
    print("PARTIAL WEIGHT LOADING SUMMARY")
    print("="*65)
    print(f"Successfully loaded keys : {len(loaded_keys)}")
    print(f"Skipped keys             : {len(skipped_keys)}")
    print(f"Missing keys             : {len(missing)}")
    print(f"Unexpected keys          : {len(unexpected)}")
    print("="*65)

    return model.to(device)


def create_hiera_model():
    model = create_reduced_hiera_with_partial_pretrained(
        num_classes=NUM_CLASSES,
        stages=(2, 3, 11, 3),
        embed_dim=96,
        device=DEVICE
    )
    freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT)
    return model


# ============================================================
# EDGE DEPLOYMENT METRICS
# ============================================================
def print_edge_deployment_metrics(model, device=DEVICE, input_shape=(1, 3, 16, 224, 224)):
    print("\n" + "=" * 70)
    print("EDGE DEPLOYMENT METRICS")
    print("=" * 70)

    model.eval()
    model.to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    param_size_mb = total_params * 4 / (1024 ** 2)
    param_size_mb_fp16 = total_params * 2 / (1024 ** 2)

    print(f"\n[1] MEMORY FOOTPRINT")
    print(f"    Total parameters      : {total_params:,}")
    print(f"    Trainable parameters  : {trainable_params:,}")
    print(f"    Model size (FP32)     : {param_size_mb:.2f} MB")
    print(f"    Model size (FP16 est.): {param_size_mb_fp16:.2f} MB")

    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
        torch.cuda.empty_cache()
        dummy = torch.randn(*input_shape, device=device)
        with torch.no_grad():
            _ = model(dummy)
        peak_mem_mb = torch.cuda.max_memory_allocated() / (1024 ** 2)
        print(f"    Peak GPU memory (fwd) : {peak_mem_mb:.2f} MB")

    print(f"\n[2] COMPUTATIONAL COMPLEXITY (FLOPs)")
    try:
        dummy = torch.randn(*input_shape, device=device)
        flops, params = profile(model, inputs=(dummy,), verbose=False)
        flops_str, params_str = clever_format([flops, params], "%.3f")
        print(f"    FLOPs                 : {flops_str}")
        print(f"    Params (thop)         : {params_str}")
    except Exception as e:
        print(f"    FLOPs calculation failed: {e}")

    print(f"\n[3] INFERENCE TIME (Latency)")
    dummy = torch.randn(*input_shape, device=device)
    with torch.no_grad():
        for _ in range(10):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()

    n_runs = 50
    start = time.perf_counter()
    with torch.no_grad():
        for _ in range(n_runs):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    end = time.perf_counter()

    avg_latency_ms = (end - start) / n_runs * 1000
    fps = 1000.0 / avg_latency_ms
    print(f"    Avg latency (batch=1) : {avg_latency_ms:.2f} ms")
    print(f"    Throughput            : {fps:.1f} FPS")

    print(f"\n[4] POWER DRAW")
    print(f"    Measure on target edge device (Jetson / mobile NPU)")

    print(f"\n[5] THERMAL LIMITS")
    print(f"    Monitor on target hardware under sustained load")

    print(f"\n[6] MODEL ACCURACY")
    print(f"    See FINAL RESULTS section after cross-validation")
    print("=" * 70 + "\n")


# ============================================================
# LOAD FULL DATASET
# ============================================================
print("\nLoading full dataset...")
full_dataset_for_split = ActivityDataset(
    DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
)


# ============================================================
# TRAIN / EVAL
# ============================================================
def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for videos, labels, _ in tqdm(loader, desc=" train", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * videos.size(0)
        preds = logits.argmax(1)
        correct += (preds == labels).sum().item()
        total += videos.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    all_preds, all_labels, all_probs, all_paths = [], [], [], []

    for videos, labels, paths in tqdm(loader, desc=" val", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        total_loss += loss.item() * videos.size(0)
        probs = torch.softmax(logits.float(), dim=1)
        preds = logits.argmax(1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
        all_paths.extend(paths)

    return (total_loss / len(loader.dataset),
            np.array(all_preds),
            np.array(all_labels),
            np.array(all_probs),
            all_paths)


def plot_cm(cm, title, normalize=True):
    if normalize:
        cm_plot = cm.astype("float") / (cm.sum(axis=1)[:, np.newaxis] + 1e-8)
        fmt = ".2f"
    else:
        cm_plot = cm
        fmt = "d"

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm_plot, annot=True, fmt=fmt, cmap="Blues",
                xticklabels=CLASSES_LIST, yticklabels=CLASSES_LIST)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# PROFILE MODEL ONCE
# ============================================================
print("\nCreating temporary model for edge metrics...")
_temp_model = create_hiera_model()
print_edge_deployment_metrics(_temp_model)
del _temp_model
torch.cuda.empty_cache()


# ============================================================
# 5-FOLD CROSS VALIDATION
# ============================================================
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)

accuracy_scores = []
top2_scores = []
f1_macro_scores = []
f1_weighted_scores = []
precision_scores = []
recall_scores = []
auc_scores = []
confusion_matrices = []

overall_best_f1 = 0.0
overall_best_path = None

fold = 1
for train_idx, val_idx in skf.split(np.zeros(len(full_dataset_for_split)), full_dataset_for_split.labels):
    print(f"\n{'='*70}")
    print(f"========== Fold {fold}/{NUM_FOLDS} ==========")
    print(f"{'='*70}")

    train_dataset = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=True)
    val_dataset   = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False)

    train_subset = Subset(train_dataset, train_idx)
    val_subset   = Subset(val_dataset, val_idx)

    train_labels = full_dataset_for_split.labels[train_idx]
    class_weights = compute_class_weight("balanced", classes=np.unique(train_labels), y=train_labels)
    class_weight_tensor = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
    print("Class weights (this fold):", dict(enumerate(np.round(class_weights, 3))))

    train_loader = DataLoader(
        train_subset, batch_size=BATCH_SIZE, shuffle=True,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )
    val_loader = DataLoader(
        val_subset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )

    model = create_hiera_model()
    criterion = nn.CrossEntropyLoss(weight=class_weight_tensor)

    backbone_params = [p for n, p in model.named_parameters() if "head" not in n and p.requires_grad]
    head_params     = [p for n, p in model.named_parameters() if "head" in n and p.requires_grad]

    optimizer = torch.optim.AdamW([
        {"params": backbone_params, "lr": LR_BACKBONE},
        {"params": head_params,     "lr": LR_HEAD}
    ], weight_decay=WEIGHT_DECAY)

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    best_f1 = 0.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        val_loss, y_pred, y_true, y_prob, _ = evaluate(model, val_loader, criterion)
        scheduler.step()

        current_lrs = [group['lr'] for group in optimizer.param_groups]
        print(f"  → LRs: backbone={current_lrs[0]:.2e} | head={current_lrs[1]:.2e}")

        val_acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average="macro", zero_division=0)

        try:
            top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
        except Exception:
            top2_acc = 0.0

        print(f"Epoch {epoch+1:02d} | "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} "
              f"top2={top2_acc:.4f} macro_F1={f1_macro:.4f}")

        if f1_macro > best_f1:
            best_f1 = f1_macro
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
            print(f" → New best macro F1: {best_f1:.4f}")
        else:
            epochs_no_improve += 1
            print(f" → No improvement for {epochs_no_improve}/{PATIENCE} epochs")

        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch+1}")
            break

    model.load_state_dict(best_state)
    model.to(DEVICE)
    val_loss, y_pred, y_true, y_prob, video_paths = evaluate(model, val_loader, criterion)

    acc = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall_macro    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_macro        = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_weighted     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    try:
        top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
    except Exception:
        top2_acc = 0.0

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc_macro = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except Exception as e:
        print(f"AUC calculation warning: {e}")
        auc_macro = 0.0

    accuracy_scores.append(acc)
    top2_scores.append(top2_acc)
    f1_macro_scores.append(f1_macro)
    f1_weighted_scores.append(f1_weighted)
    precision_scores.append(precision_macro)
    recall_scores.append(recall_macro)
    auc_scores.append(auc_macro)

    cm = confusion_matrix(y_true, y_pred)
    confusion_matrices.append(cm)
    plot_cm(cm, f"Activity Recognition CM – Fold {fold}")

    print(f"\n{'='*60}")
    print(f"Fold {fold} Final Metrics:")
    print(f"{'='*60}")
    print(f"Accuracy        : {acc:.4f}")
    print(f"Top-2 Accuracy  : {top2_acc:.4f}")
    print(f"Precision (Macro): {precision_macro:.4f}")
    print(f"Recall (Macro)  : {recall_macro:.4f}")
    print(f"F1-Score (Macro): {f1_macro:.4f}")
    print(f"F1-Score (Weighted): {f1_weighted:.4f}")
    print(f"AUC (Macro OVR) : {auc_macro:.4f}")
    print(f"{'='*60}")

    print(f"\nFold {fold} Classification Report:")
    print(classification_report(y_true, y_pred, target_names=CLASSES_LIST, digits=4, zero_division=0))

    true_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_true]
    pred_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_pred]

    df = pd.DataFrame({
        "Video_Path": video_paths,
        "True_Class": true_class_names,
        "Predicted_Class": pred_class_names,
        "Correct": [t == p for t, p in zip(true_class_names, pred_class_names)]
    })

    excel_path = os.path.join(SAVE_DIR, f"predictions_fold_{fold}.xlsx")
    df.to_excel(excel_path, index=False)
    print(f"Excel saved → {excel_path}")

    fold_save_path = os.path.join(SAVE_DIR, f"best_hiera_modified_fold_{fold}.pth")
    torch.save({
        "model_state_dict": best_state,
        "fold": fold,
        "macro_f1": best_f1,
        "accuracy": acc,
        "top2_accuracy": top2_acc,
        "class_to_idx": full_dataset_for_split.class_to_idx,
        "idx_to_class": full_dataset_for_split.idx_to_class,
        "classes": CLASSES_LIST,
        "stages": (2, 3, 11, 3),
    }, fold_save_path)
    print(f"Model saved → {fold_save_path}")

    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1
        overall_best_path = fold_save_path

    del optimizer, scaler, model
    torch.cuda.empty_cache()
    fold += 1


# ============================================================
# FINAL RESULTS
# ============================================================
print("\n" + "="*70)
print("FINAL RESULTS (Modified Hiera - stages=(2,3,11,3) + Partial Pretrained + FULL TRAINING)")
print("="*70)

print(f"\nPer-fold Accuracy : {[f'{x:.4f}' for x in accuracy_scores]}")
print(f"Mean ± std Accuracy : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")

print(f"\nPer-fold Top-2 Accuracy : {[f'{x:.4f}' for x in top2_scores]}")
print(f"Mean ± std Top-2 : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")

print(f"\nPer-fold Precision : {[f'{x:.4f}' for x in precision_scores]}")
print(f"Mean ± std Precision : {np.mean(precision_scores):.4f} ± {np.std(precision_scores):.4f}")

print(f"\nPer-fold Recall : {[f'{x:.4f}' for x in recall_scores]}")
print(f"Mean ± std Recall : {np.mean(recall_scores):.4f} ± {np.std(recall_scores):.4f}")

print(f"\nPer-fold Macro F1 : {[f'{x:.4f}' for x in f1_macro_scores]}")
print(f"Mean ± std Macro F1 : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")

print(f"\nPer-fold Weighted F1 : {[f'{x:.4f}' for x in f1_weighted_scores]}")
print(f"Mean ± std Weighted F1 : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")

print(f"\nPer-fold AUC (Macro) : {[f'{x:.4f}' for x in auc_scores]}")
print(f"Mean ± std AUC : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")

conf_sum = np.sum(confusion_matrices, axis=0)
plot_cm(conf_sum, "Aggregated Confusion Matrix (All Folds)", normalize=False)
plot_cm(conf_sum, "Aggregated Confusion Matrix (Normalized)", normalize=True)

print(f"\nOverall best model (Macro F1={overall_best_f1:.4f}) saved at:")
print(overall_best_path)

print("\n" + "=" * 70)
print("EDGE DEPLOYMENT SUMMARY (with measured accuracy)")
print("=" * 70)
print(f"Model Architecture     : Modified Hiera stages=(2,3,11,3)")
print(f"Pretrained Loading     : Partial (from original Hiera-Base)")
print(f"Trainable Strategy     : FULL TRAINING (all blocks + Head)")
print(f"Mean Accuracy          : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")
print(f"Mean Top-2 Accuracy    : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")
print(f"Mean Macro F1          : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")
print(f"Mean Weighted F1       : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")
print(f"Mean AUC               : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("=" * 70)
```

```python
# ============================================================
# Toddler Activity Recognition - Modified Hiera
# batch size 4
# Reduced stages=(2,3,9,3) + Partial Pretrained Loading + FULL TRAINING
# + Class weights inside each fold
# + Consistent Video Augmentations
# + LR logging + Top-2 Accuracy + Edge Deployment Metrics
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

!pip install -q torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
!pip install -q hiera-transformer av opencv-python tqdm scikit-learn seaborn pandas openpyxl thop

import os
import random
import time
from pathlib import Path
from collections import Counter, OrderedDict

import av
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.transforms import functional as TF
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    precision_score, recall_score, roc_auc_score, top_k_accuracy_score
)
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns
import hiera
from thop import profile, clever_format

# ────────────────────────────────────────────────
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    torch.backends.cudnn.benchmark = True

# ===================== Settings =====================
IMAGE_HEIGHT, IMAGE_WIDTH = 224, 224
SEQUENCE_LENGTH = 16
BATCH_SIZE = 4
EPOCHS = 25
LR_BACKBONE = 1e-5
LR_HEAD = 5e-5
WEIGHT_DECAY = 0.05
NUM_FOLDS = 5
NUM_WORKERS = 4
PATIENCE = 4
PIN_MEMORY = True
USE_AMP = True
TRAIN_LAST_PERCENT = 1.0          # FULL TRAINING (all blocks + head)

DATA_DIR = "/content/drive/MyDrive/DatasetFinal"
SAVE_DIR = "hiera_modified_stages2393_full_train"
os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 16 Classes =====================
CLASSES_LIST = [
    'Climb_Down_Stairs', 'Climb_Up_Stairs', 'Crawl', 'Drink', 'Eat',
    'Emesis', 'Exit_Crib', 'Fall_Down', 'Hold_Knife', 'Play_Slide',
    'Play_Stacking_Rings', 'Play_Swing', 'Play_Walker', 'Sleep', 'W_Sit', 'Walk'
]
NUM_CLASSES = len(CLASSES_LIST)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 70)
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
print(f"Strategy: Reduced Hiera (2,3,9,3) + Partial Pretrained + FULL TRAINING")
print("=" * 70)


# ============================================================
# VIDEO AUGMENTATION
# ============================================================
class VideoAugmentation:
    def __init__(self, size=224, is_train=True):
        self.size = size
        self.is_train = is_train
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

    def __call__(self, frames):
        pil_frames = [transforms.ToPILImage()(f) for f in frames]

        if self.is_train:
            if random.random() < 0.5:
                pil_frames = [TF.hflip(f) for f in pil_frames]

            brightness = random.uniform(0.8, 1.2)
            contrast   = random.uniform(0.8, 1.2)
            saturation = random.uniform(0.8, 1.2)
            hue        = random.uniform(-0.05, 0.05)

            pil_frames = [TF.adjust_brightness(f, brightness) for f in pil_frames]
            pil_frames = [TF.adjust_contrast(f, contrast) for f in pil_frames]
            pil_frames = [TF.adjust_saturation(f, saturation) for f in pil_frames]
            pil_frames = [TF.adjust_hue(f, hue) for f in pil_frames]

        tensors = []
        for f in pil_frames:
            f = TF.resize(f, [self.size, self.size])
            t = TF.to_tensor(f)
            if self.is_train and random.random() < 0.3:
                noise = torch.randn_like(t) * 0.02
                t = torch.clamp(t + noise, 0.0, 1.0)
            t = self.normalize(t)
            tensors.append(t)

        return torch.stack(tensors)


def load_video_frames_raw(video_path, num_frames=16):
    container = av.open(video_path)
    frames = []
    for frame in container.decode(video=0):
        img = frame.to_ndarray(format="rgb24")
        frames.append(img)
    container.close()

    if len(frames) == 0:
        return None

    indices = np.linspace(0, len(frames) - 1, num_frames).astype(int)
    return [frames[i] for i in indices]


# ============================================================
# DATASET
# ============================================================
class ActivityDataset(Dataset):
    def __init__(self, data_dir, classes_list, num_frames=16, size=224, is_train=True):
        self.samples = []
        self.labels = []
        self.num_frames = num_frames
        self.size = size
        self.is_train = is_train
        self.class_to_idx = {cls: idx for idx, cls in enumerate(classes_list)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}
        self.augment = VideoAugmentation(size=size, is_train=is_train)

        for class_name in classes_list:
            class_path = Path(data_dir) / class_name
            if not class_path.is_dir():
                print(f"Warning: folder not found → {class_path}")
                continue
            count = 0
            for video_file in sorted(class_path.glob("*")):
                if video_file.suffix.lower() in {".mp4", ".avi", ".mov", ".mkv"}:
                    self.samples.append(str(video_file))
                    self.labels.append(self.class_to_idx[class_name])
                    count += 1
            print(f"Loaded {count} videos from '{class_name}'")

        self.labels = np.array(self.labels)
        print(f"\nTotal videos: {len(self.samples)}")
        print("Class distribution:", dict(Counter(self.labels)))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path = self.samples[idx]
        label = self.labels[idx]
        frames = load_video_frames_raw(path, self.num_frames)

        if frames is None:
            frames = [np.zeros((self.size, self.size, 3), dtype=np.uint8) for _ in range(self.num_frames)]

        video_tensor = self.augment(frames)
        return video_tensor, label, path


def collate_fn(batch):
    videos, labels, paths = zip(*batch)
    videos = torch.stack(videos)
    labels = torch.tensor(labels, dtype=torch.long)
    return videos, labels, list(paths)


# ============================================================
# FULL TRAINING (no freezing)
# ============================================================
def freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT):
    """
    Full training mode: every parameter is trainable.
    The train_last_percent argument is kept only for API compatibility.
    """
    for param in model.parameters():
        param.requires_grad = True

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"\n[Full Training] Trainable params: {trainable:,} / {total:,} "
          f"({100 * trainable / total:.1f}%)")


# ============================================================
# PARTIAL WEIGHT LOADING + REDUCED MODEL
# ============================================================
def create_reduced_hiera_with_partial_pretrained(
    num_classes=16,
    stages=(2, 3, 9, 3),
    embed_dim=96,
    device="cuda"
):
    model = hiera.Hiera(
        input_size=(16, 224, 224),
        in_chans=3,
        embed_dim=embed_dim,
        num_heads=1,
        num_classes=num_classes,
        stages=stages,
        q_pool=3,
        q_stride=(1, 2, 2),
        mask_unit_size=(1, 8, 8),
        patch_kernel=(3, 7, 7),
        patch_stride=(2, 4, 4),
        patch_padding=(1, 3, 3),
        sep_pos_embed=True,
        drop_path_rate=0.15,
    )

    print("\nLoading original pretrained Hiera-Base weights...")
    pretrained = hiera.hiera_base_16x224(
        pretrained=True,
        checkpoint="mae_k400_ft_k400",
        num_classes=400
    )
    pretrained_dict = pretrained.state_dict()
    model_dict = model.state_dict()

    new_state_dict = OrderedDict()
    loaded_keys = []
    skipped_keys = []

    # Original Hiera-Base stages ≈ (2, 3, 16, 3) → 24 blocks
    # Target stages = (2, 3, 9, 3) → 17 blocks
    # Mapping:
    #   blocks 0-4   (stage1+2)   → keep as-is
    #   blocks 5-13  (9 of stage3) → keep
    #   blocks 14-20 (remaining stage3) → skip
    #   blocks 21-23 (stage4) → shift by -7 → 14-16
    for key, value in pretrained_dict.items():
        if "head" in key:
            skipped_keys.append(key)
            continue

        if "blocks" in key:
            parts = key.split(".")
            block_idx = int(parts[1])

            if block_idx <= 4:
                new_block_idx = block_idx
            elif 5 <= block_idx <= 13:
                new_block_idx = block_idx
            elif block_idx >= 21:
                new_block_idx = block_idx - 7
            else:
                skipped_keys.append(key)
                continue

            new_key = key.replace(f"blocks.{block_idx}.", f"blocks.{new_block_idx}.")

            if new_key in model_dict and model_dict[new_key].shape == value.shape:
                new_state_dict[new_key] = value
                loaded_keys.append(new_key)
            else:
                skipped_keys.append(key)
        else:
            if key in model_dict and model_dict[key].shape == value.shape:
                new_state_dict[key] = value
                loaded_keys.append(key)
            else:
                skipped_keys.append(key)

    missing, unexpected = model.load_state_dict(new_state_dict, strict=False)

    print("\n" + "="*65)
    print("PARTIAL WEIGHT LOADING SUMMARY")
    print("="*65)
    print(f"Successfully loaded keys : {len(loaded_keys)}")
    print(f"Skipped keys             : {len(skipped_keys)}")
    print(f"Missing keys             : {len(missing)}")
    print(f"Unexpected keys          : {len(unexpected)}")
    print("="*65)

    return model.to(device)


def create_hiera_model():
    model = create_reduced_hiera_with_partial_pretrained(
        num_classes=NUM_CLASSES,
        stages=(2, 3, 9, 3),
        embed_dim=96,
        device=DEVICE
    )
    freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT)
    return model


# ============================================================
# EDGE DEPLOYMENT METRICS
# ============================================================
def print_edge_deployment_metrics(model, device=DEVICE, input_shape=(1, 3, 16, 224, 224)):
    print("\n" + "=" * 70)
    print("EDGE DEPLOYMENT METRICS")
    print("=" * 70)

    model.eval()
    model.to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    param_size_mb = total_params * 4 / (1024 ** 2)
    param_size_mb_fp16 = total_params * 2 / (1024 ** 2)

    print(f"\n[1] MEMORY FOOTPRINT")
    print(f"    Total parameters      : {total_params:,}")
    print(f"    Trainable parameters  : {trainable_params:,}")
    print(f"    Model size (FP32)     : {param_size_mb:.2f} MB")
    print(f"    Model size (FP16 est.): {param_size_mb_fp16:.2f} MB")

    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
        torch.cuda.empty_cache()
        dummy = torch.randn(*input_shape, device=device)
        with torch.no_grad():
            _ = model(dummy)
        peak_mem_mb = torch.cuda.max_memory_allocated() / (1024 ** 2)
        print(f"    Peak GPU memory (fwd) : {peak_mem_mb:.2f} MB")

    print(f"\n[2] COMPUTATIONAL COMPLEXITY (FLOPs)")
    try:
        dummy = torch.randn(*input_shape, device=device)
        flops, params = profile(model, inputs=(dummy,), verbose=False)
        flops_str, params_str = clever_format([flops, params], "%.3f")
        print(f"    FLOPs                 : {flops_str}")
        print(f"    Params (thop)         : {params_str}")
    except Exception as e:
        print(f"    FLOPs calculation failed: {e}")

    print(f"\n[3] INFERENCE TIME (Latency)")
    dummy = torch.randn(*input_shape, device=device)
    with torch.no_grad():
        for _ in range(10):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()

    n_runs = 50
    start = time.perf_counter()
    with torch.no_grad():
        for _ in range(n_runs):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    end = time.perf_counter()

    avg_latency_ms = (end - start) / n_runs * 1000
    fps = 1000.0 / avg_latency_ms
    print(f"    Avg latency (batch=1) : {avg_latency_ms:.2f} ms")
    print(f"    Throughput            : {fps:.1f} FPS")

    print(f"\n[4] POWER DRAW")
    print(f"    Measure on target edge device (Jetson / mobile NPU)")

    print(f"\n[5] THERMAL LIMITS")
    print(f"    Monitor on target hardware under sustained load")

    print(f"\n[6] MODEL ACCURACY")
    print(f"    See FINAL RESULTS section after cross-validation")
    print("=" * 70 + "\n")


# ============================================================
# LOAD FULL DATASET
# ============================================================
print("\nLoading full dataset...")
full_dataset_for_split = ActivityDataset(
    DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
)


# ============================================================
# TRAIN / EVAL
# ============================================================
def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for videos, labels, _ in tqdm(loader, desc=" train", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * videos.size(0)
        preds = logits.argmax(1)
        correct += (preds == labels).sum().item()
        total += videos.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    all_preds, all_labels, all_probs, all_paths = [], [], [], []

    for videos, labels, paths in tqdm(loader, desc=" val", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        total_loss += loss.item() * videos.size(0)
        probs = torch.softmax(logits.float(), dim=1)
        preds = logits.argmax(1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
        all_paths.extend(paths)

    return (total_loss / len(loader.dataset),
            np.array(all_preds),
            np.array(all_labels),
            np.array(all_probs),
            all_paths)


def plot_cm(cm, title, normalize=True):
    if normalize:
        cm_plot = cm.astype("float") / (cm.sum(axis=1)[:, np.newaxis] + 1e-8)
        fmt = ".2f"
    else:
        cm_plot = cm
        fmt = "d"

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm_plot, annot=True, fmt=fmt, cmap="Blues",
                xticklabels=CLASSES_LIST, yticklabels=CLASSES_LIST)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# PROFILE MODEL ONCE
# ============================================================
print("\nCreating temporary model for edge metrics...")
_temp_model = create_hiera_model()
print_edge_deployment_metrics(_temp_model)
del _temp_model
torch.cuda.empty_cache()


# ============================================================
# 5-FOLD CROSS VALIDATION
# ============================================================
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)

accuracy_scores = []
top2_scores = []
f1_macro_scores = []
f1_weighted_scores = []
precision_scores = []
recall_scores = []
auc_scores = []
confusion_matrices = []

overall_best_f1 = 0.0
overall_best_path = None

fold = 1
for train_idx, val_idx in skf.split(np.zeros(len(full_dataset_for_split)), full_dataset_for_split.labels):
    print(f"\n{'='*70}")
    print(f"========== Fold {fold}/{NUM_FOLDS} ==========")
    print(f"{'='*70}")

    train_dataset = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=True)
    val_dataset   = ActivityDataset(DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False)

    train_subset = Subset(train_dataset, train_idx)
    val_subset   = Subset(val_dataset, val_idx)

    train_labels = full_dataset_for_split.labels[train_idx]
    class_weights = compute_class_weight("balanced", classes=np.unique(train_labels), y=train_labels)
    class_weight_tensor = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
    print("Class weights (this fold):", dict(enumerate(np.round(class_weights, 3))))

    train_loader = DataLoader(
        train_subset, batch_size=BATCH_SIZE, shuffle=True,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )
    val_loader = DataLoader(
        val_subset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )

    model = create_hiera_model()
    criterion = nn.CrossEntropyLoss(weight=class_weight_tensor)

    backbone_params = [p for n, p in model.named_parameters() if "head" not in n and p.requires_grad]
    head_params     = [p for n, p in model.named_parameters() if "head" in n and p.requires_grad]

    optimizer = torch.optim.AdamW([
        {"params": backbone_params, "lr": LR_BACKBONE},
        {"params": head_params,     "lr": LR_HEAD}
    ], weight_decay=WEIGHT_DECAY)

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    best_f1 = 0.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        val_loss, y_pred, y_true, y_prob, _ = evaluate(model, val_loader, criterion)
        scheduler.step()

        current_lrs = [group['lr'] for group in optimizer.param_groups]
        print(f"  → LRs: backbone={current_lrs[0]:.2e} | head={current_lrs[1]:.2e}")

        val_acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average="macro", zero_division=0)

        try:
            top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
        except Exception:
            top2_acc = 0.0

        print(f"Epoch {epoch+1:02d} | "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} "
              f"top2={top2_acc:.4f} macro_F1={f1_macro:.4f}")

        if f1_macro > best_f1:
            best_f1 = f1_macro
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
            print(f" → New best macro F1: {best_f1:.4f}")
        else:
            epochs_no_improve += 1
            print(f" → No improvement for {epochs_no_improve}/{PATIENCE} epochs")

        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch+1}")
            break

    model.load_state_dict(best_state)
    model.to(DEVICE)
    val_loss, y_pred, y_true, y_prob, video_paths = evaluate(model, val_loader, criterion)

    acc = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall_macro    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_macro        = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_weighted     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    try:
        top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
    except Exception:
        top2_acc = 0.0

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc_macro = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except Exception as e:
        print(f"AUC calculation warning: {e}")
        auc_macro = 0.0

    accuracy_scores.append(acc)
    top2_scores.append(top2_acc)
    f1_macro_scores.append(f1_macro)
    f1_weighted_scores.append(f1_weighted)
    precision_scores.append(precision_macro)
    recall_scores.append(recall_macro)
    auc_scores.append(auc_macro)

    cm = confusion_matrix(y_true, y_pred)
    confusion_matrices.append(cm)
    plot_cm(cm, f"Activity Recognition CM – Fold {fold}")

    print(f"\n{'='*60}")
    print(f"Fold {fold} Final Metrics:")
    print(f"{'='*60}")
    print(f"Accuracy        : {acc:.4f}")
    print(f"Top-2 Accuracy  : {top2_acc:.4f}")
    print(f"Precision (Macro): {precision_macro:.4f}")
    print(f"Recall (Macro)  : {recall_macro:.4f}")
    print(f"F1-Score (Macro): {f1_macro:.4f}")
    print(f"F1-Score (Weighted): {f1_weighted:.4f}")
    print(f"AUC (Macro OVR) : {auc_macro:.4f}")
    print(f"{'='*60}")

    print(f"\nFold {fold} Classification Report:")
    print(classification_report(y_true, y_pred, target_names=CLASSES_LIST, digits=4, zero_division=0))

    true_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_true]
    pred_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_pred]

    df = pd.DataFrame({
        "Video_Path": video_paths,
        "True_Class": true_class_names,
        "Predicted_Class": pred_class_names,
        "Correct": [t == p for t, p in zip(true_class_names, pred_class_names)]
    })

    excel_path = os.path.join(SAVE_DIR, f"predictions_fold_{fold}.xlsx")
    df.to_excel(excel_path, index=False)
    print(f"Excel saved → {excel_path}")

    fold_save_path = os.path.join(SAVE_DIR, f"best_hiera_modified_fold_{fold}.pth")
    torch.save({
        "model_state_dict": best_state,
        "fold": fold,
        "macro_f1": best_f1,
        "accuracy": acc,
        "top2_accuracy": top2_acc,
        "class_to_idx": full_dataset_for_split.class_to_idx,
        "idx_to_class": full_dataset_for_split.idx_to_class,
        "classes": CLASSES_LIST,
        "stages": (2, 3, 9, 3),
    }, fold_save_path)
    print(f"Model saved → {fold_save_path}")

    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1
        overall_best_path = fold_save_path

    del optimizer, scaler, model
    torch.cuda.empty_cache()
    fold += 1


# ============================================================
# FINAL RESULTS
# ============================================================
print("\n" + "="*70)
print("FINAL RESULTS (Modified Hiera - stages=(2,3,9,3) + Partial Pretrained + FULL TRAINING)")
print("="*70)

print(f"\nPer-fold Accuracy : {[f'{x:.4f}' for x in accuracy_scores]}")
print(f"Mean ± std Accuracy : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")

print(f"\nPer-fold Top-2 Accuracy : {[f'{x:.4f}' for x in top2_scores]}")
print(f"Mean ± std Top-2 : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")

print(f"\nPer-fold Precision : {[f'{x:.4f}' for x in precision_scores]}")
print(f"Mean ± std Precision : {np.mean(precision_scores):.4f} ± {np.std(precision_scores):.4f}")

print(f"\nPer-fold Recall : {[f'{x:.4f}' for x in recall_scores]}")
print(f"Mean ± std Recall : {np.mean(recall_scores):.4f} ± {np.std(recall_scores):.4f}")

print(f"\nPer-fold Macro F1 : {[f'{x:.4f}' for x in f1_macro_scores]}")
print(f"Mean ± std Macro F1 : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")

print(f"\nPer-fold Weighted F1 : {[f'{x:.4f}' for x in f1_weighted_scores]}")
print(f"Mean ± std Weighted F1 : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")

print(f"\nPer-fold AUC (Macro) : {[f'{x:.4f}' for x in auc_scores]}")
print(f"Mean ± std AUC : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")

conf_sum = np.sum(confusion_matrices, axis=0)
plot_cm(conf_sum, "Aggregated Confusion Matrix (All Folds)", normalize=False)
plot_cm(conf_sum, "Aggregated Confusion Matrix (Normalized)", normalize=True)

print(f"\nOverall best model (Macro F1={overall_best_f1:.4f}) saved at:")
print(overall_best_path)

print("\n" + "=" * 70)
print("EDGE DEPLOYMENT SUMMARY (with measured accuracy)")
print("=" * 70)
print(f"Model Architecture     : Modified Hiera stages=(2,3,9,3)")
print(f"Pretrained Loading     : Partial (from original Hiera-Base)")
print(f"Trainable Strategy     : FULL TRAINING (all blocks + Head)")
print(f"Mean Accuracy          : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")
print(f"Mean Top-2 Accuracy    : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")
print(f"Mean Macro F1          : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")
print(f"Mean Weighted F1       : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")
print(f"Mean AUC               : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("=" * 70)
```
