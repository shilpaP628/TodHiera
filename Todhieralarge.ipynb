```python
# ============================================================
# Toddler Activity Recognition with Hiera
# 16 Classes + Last 50% Blocks Trainable + L4 Optimized
# + Class weights computed inside each fold
# + Consistent Video Augmentations (HFlip + ColorJitter + Gaussian Noise)
# + LR logging + Top-2 Accuracy
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

!pip install -q torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
!pip install -q hiera-transformer av opencv-python tqdm scikit-learn seaborn pandas openpyxl

import os
import random
from pathlib import Path
from collections import Counter

import av
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.transforms import functional as TF
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    precision_score, recall_score, roc_auc_score, top_k_accuracy_score
)
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns
import hiera

# ────────────────────────────────────────────────
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    torch.backends.cudnn.benchmark = True

# ===================== L4 Optimized Settings =====================
IMAGE_HEIGHT, IMAGE_WIDTH = 224, 224
SEQUENCE_LENGTH = 16
BATCH_SIZE = 4
EPOCHS = 25
LR_BACKBONE = 1e-5          # for the unfrozen blocks
LR_HEAD = 5e-5              # for the head
WEIGHT_DECAY = 0.05
NUM_FOLDS = 5
NUM_WORKERS = 4
PATIENCE = 4
PIN_MEMORY = True
USE_AMP = True
TRAIN_LAST_PERCENT = 0.50   # Last 50% of blocks

DATA_DIR = "/content/drive/MyDrive/DatasetFinal"
SAVE_DIR = "hiera_activity_checkpoints_16cls_last50pct_L4_aug"
os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 16 Classes =====================
CLASSES_LIST = [
    'Climb_Down_Stairs',
    'Climb_Up_Stairs',
    'Crawl',
    'Drink',
    'Eat',
    'Emesis',
    'Exit_Crib',
    'Fall_Down',
    'Hold_Knife',
    'Play_Slide',
    'Play_Stacking_Rings',
    'Play_Swing',
    'Play_Walker',
    'Sleep',
    'W_Sit',
    'Walk'
]
NUM_CLASSES = len(CLASSES_LIST)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 70)
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
    print("GPU Memory:", round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 1), "GB")
print("Classes:", CLASSES_LIST)
print(f"Number of classes: {NUM_CLASSES}")
print(f"Strategy: Last {TRAIN_LAST_PERCENT*100:.0f}% blocks trainable + Head")
print(f"AMP: {USE_AMP} | Workers: {NUM_WORKERS} | Pin Memory: {PIN_MEMORY}")
print("=" * 70)


# ============================================================
# VIDEO AUGMENTATION (consistent across all frames of a clip)
# ============================================================
class VideoAugmentation:
    def __init__(self, size=224, is_train=True):
        self.size = size
        self.is_train = is_train
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

    def __call__(self, frames):
        """
        frames: list of numpy arrays (H, W, C) in RGB
        Returns: Tensor of shape (T, C, H, W)
        """
        # Convert to PIL
        pil_frames = [transforms.ToPILImage()(f) for f in frames]

        # ---------- Train-time augmentations (identical params for every frame) ----------
        if self.is_train:
            # 1. Horizontal Flip
            if random.random() < 0.5:
                pil_frames = [TF.hflip(f) for f in pil_frames]

            # 2. Color Jitter (same parameters for the whole clip)
            brightness = random.uniform(0.8, 1.2)
            contrast   = random.uniform(0.8, 1.2)
            saturation = random.uniform(0.8, 1.2)
            hue        = random.uniform(-0.05, 0.05)

            pil_frames = [TF.adjust_brightness(f, brightness) for f in pil_frames]
            pil_frames = [TF.adjust_contrast(f, contrast)     for f in pil_frames]
            pil_frames = [TF.adjust_saturation(f, saturation) for f in pil_frames]
            pil_frames = [TF.adjust_hue(f, hue)               for f in pil_frames]

        # Resize + ToTensor + optional Gaussian noise
        tensors = []
        for f in pil_frames:
            f = TF.resize(f, [self.size, self.size])
            t = TF.to_tensor(f)          # (C, H, W) in [0, 1]

            # 3. Gaussian Noise (train only)
            if self.is_train and random.random() < 0.3:
                noise = torch.randn_like(t) * 0.02
                t = torch.clamp(t + noise, 0.0, 1.0)

            t = self.normalize(t)
            tensors.append(t)

        return torch.stack(tensors)      # (T, C, H, W)


# ============================================================
# FRAME EXTRACTION (raw frames only)
# ============================================================
def load_video_frames_raw(video_path, num_frames=16):
    container = av.open(video_path)
    frames = []
    for frame in container.decode(video=0):
        img = frame.to_ndarray(format="rgb24")
        frames.append(img)
    container.close()

    if len(frames) == 0:
        return None

    indices = np.linspace(0, len(frames) - 1, num_frames).astype(int)
    frames = [frames[i] for i in indices]
    return frames


# ============================================================
# DATASET
# ============================================================
class ActivityDataset(Dataset):
    def __init__(self, data_dir, classes_list, num_frames=16, size=224, is_train=True):
        self.samples = []
        self.labels = []
        self.num_frames = num_frames
        self.size = size
        self.is_train = is_train
        self.class_to_idx = {cls: idx for idx, cls in enumerate(classes_list)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}
        self.augment = VideoAugmentation(size=size, is_train=is_train)

        for class_name in classes_list:
            class_path = Path(data_dir) / class_name
            if not class_path.is_dir():
                print(f"Warning: folder not found → {class_path}")
                continue

            count = 0
            for video_file in sorted(class_path.glob("*")):
                if video_file.suffix.lower() in {".mp4", ".avi", ".mov", ".mkv"}:
                    self.samples.append(str(video_file))
                    self.labels.append(self.class_to_idx[class_name])
                    count += 1
            print(f"Loaded {count} videos from '{class_name}'")

        self.labels = np.array(self.labels)
        print(f"\nTotal videos: {len(self.samples)}")
        print("Class distribution:", dict(Counter(self.labels)))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path = self.samples[idx]
        label = self.labels[idx]
        frames = load_video_frames_raw(path, self.num_frames)

        if frames is None:
            # fallback black clip
            frames = [np.zeros((self.size, self.size, 3), dtype=np.uint8)
                      for _ in range(self.num_frames)]

        video_tensor = self.augment(frames)   # (T, C, H, W)
        return video_tensor, label, path


def collate_fn(batch):
    videos, labels, paths = zip(*batch)
    videos = torch.stack(videos)
    labels = torch.tensor(labels, dtype=torch.long)
    return videos, labels, list(paths)


# ============================================================
# FREEZE EARLY BLOCKS - Train last 50%
# ============================================================
def freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT):
    # 1. Freeze everything
    for param in model.parameters():
        param.requires_grad = False

    # 2. Always train the head
    for param in model.head.parameters():
        param.requires_grad = True

    # 3. Unfreeze last N% of blocks
    if hasattr(model, "blocks"):
        blocks = model.blocks
        num_blocks = len(blocks)
        num_to_train = max(1, int(num_blocks * train_last_percent))
        start_idx = num_blocks - num_to_train

        print(f"\n[Freeze Strategy] Total blocks: {num_blocks}")
        print(f"[Freeze Strategy] Training last {num_to_train} blocks "
              f"({train_last_percent*100:.0f}%) → blocks[{start_idx}:]")

        for i in range(start_idx, num_blocks):
            for param in blocks[i].parameters():
                param.requires_grad = True
    else:
        print("[Warning] model.blocks not found. Unfreezing all parameters.")
        for param in model.parameters():
            param.requires_grad = True

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"[Freeze Strategy] Trainable params: {trainable:,} / {total:,} "
          f"({100 * trainable / total:.1f}%)")


# ============================================================
# LOAD FULL DATASET (labels only – we create train/val subsets later)
# ============================================================
print("\nLoading full dataset (for labels & paths)...")
# We only need the labels and paths; actual tensors are created on-the-fly
# with different augmentations for train vs val.
full_dataset_for_split = ActivityDataset(
    DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
)


# ============================================================
# MODEL
# ============================================================
def create_hiera_model():
    model = hiera.hiera_base_plus_16x224(
        pretrained=True,
        checkpoint="mae_k400_ft_k400",
        num_classes=400,
    )
    in_features = model.head.projection.in_features
    model.head.projection = nn.Linear(in_features, NUM_CLASSES)

    # Unfreeze last 50% blocks + head
    freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT)

    return model.to(DEVICE)


# ============================================================
# TRAIN / EVAL (with AMP + Top-2)
# ============================================================
def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for videos, labels, _ in tqdm(loader, desc=" train", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()   # (B, C, T, H, W)

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * videos.size(0)
        preds = logits.argmax(1)
        correct += (preds == labels).sum().item()
        total += videos.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    all_preds, all_labels, all_probs, all_paths = [], [], [], []

    for videos, labels, paths in tqdm(loader, desc=" val", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        total_loss += loss.item() * videos.size(0)
        probs = torch.softmax(logits.float(), dim=1)
        preds = logits.argmax(1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
        all_paths.extend(paths)

    return (total_loss / len(loader.dataset),
            np.array(all_preds),
            np.array(all_labels),
            np.array(all_probs),
            all_paths)


def plot_cm(cm, title, normalize=True):
    if normalize:
        cm_plot = cm.astype("float") / (cm.sum(axis=1)[:, np.newaxis] + 1e-8)
        fmt = ".2f"
    else:
        cm_plot = cm
        fmt = "d"

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm_plot, annot=True, fmt=fmt, cmap="Blues",
                xticklabels=CLASSES_LIST, yticklabels=CLASSES_LIST)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# 5-FOLD STRATIFIED CROSS VALIDATION
# ============================================================
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)

accuracy_scores = []
top2_scores = []
f1_macro_scores = []
f1_weighted_scores = []
precision_scores = []
recall_scores = []
auc_scores = []
confusion_matrices = []

overall_best_f1 = 0.0
overall_best_path = None

fold = 1
for train_idx, val_idx in skf.split(np.zeros(len(full_dataset_for_split)), full_dataset_for_split.labels):
    print(f"\n{'='*70}")
    print(f"========== Fold {fold}/{NUM_FOLDS} ==========")
    print(f"{'='*70}")

    # ---------- Create train / val datasets with correct augmentation ----------
    train_dataset = ActivityDataset(
        DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=True
    )
    val_dataset = ActivityDataset(
        DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
    )

    train_subset = Subset(train_dataset, train_idx)
    val_subset   = Subset(val_dataset,   val_idx)

    # ---------- Compute class weights using ONLY training labels ----------
    train_labels = full_dataset_for_split.labels[train_idx]
    class_weights = compute_class_weight(
        "balanced",
        classes=np.unique(train_labels),
        y=train_labels
    )
    class_weight_tensor = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
    print("Class weights (this fold):", dict(enumerate(np.round(class_weights, 3))))

    train_loader = DataLoader(
        train_subset, batch_size=BATCH_SIZE, shuffle=True,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )
    val_loader = DataLoader(
        val_subset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )

    model = create_hiera_model()
    criterion = nn.CrossEntropyLoss(weight=class_weight_tensor)

    # Differential learning rates
    backbone_params = [p for n, p in model.named_parameters() if "head" not in n and p.requires_grad]
    head_params     = [p for n, p in model.named_parameters() if "head" in n and p.requires_grad]

    optimizer = torch.optim.AdamW([
        {"params": backbone_params, "lr": LR_BACKBONE},
        {"params": head_params,     "lr": LR_HEAD}
    ], weight_decay=WEIGHT_DECAY)

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    best_f1 = 0.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        val_loss, y_pred, y_true, y_prob, _ = evaluate(model, val_loader, criterion)
        scheduler.step()

        # ---------- LR logging ----------
        current_lrs = [group['lr'] for group in optimizer.param_groups]
        print(f"  → LRs: backbone={current_lrs[0]:.2e} | head={current_lrs[1]:.2e}")

        val_acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average="macro", zero_division=0)

        # Top-2 accuracy
        try:
            top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
        except Exception:
            top2_acc = 0.0

        print(f"Epoch {epoch+1:02d} | "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} "
              f"top2={top2_acc:.4f} macro_F1={f1_macro:.4f}")

        if f1_macro > best_f1:
            best_f1 = f1_macro
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
            print(f" → New best macro F1: {best_f1:.4f}")
        else:
            epochs_no_improve += 1
            print(f" → No improvement for {epochs_no_improve}/{PATIENCE} epochs")

        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch+1}")
            break

    # ---------- Final evaluation with best weights ----------
    model.load_state_dict(best_state)
    model.to(DEVICE)
    val_loss, y_pred, y_true, y_prob, video_paths = evaluate(model, val_loader, criterion)

    acc = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall_macro    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_macro        = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_weighted     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    try:
        top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
    except Exception:
        top2_acc = 0.0

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc_macro = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except Exception as e:
        print(f"AUC calculation warning: {e}")
        auc_macro = 0.0

    accuracy_scores.append(acc)
    top2_scores.append(top2_acc)
    f1_macro_scores.append(f1_macro)
    f1_weighted_scores.append(f1_weighted)
    precision_scores.append(precision_macro)
    recall_scores.append(recall_macro)
    auc_scores.append(auc_macro)

    cm = confusion_matrix(y_true, y_pred)
    confusion_matrices.append(cm)
    plot_cm(cm, f"Activity Recognition CM – Fold {fold}")

    print(f"\n{'='*60}")
    print(f"Fold {fold} Final Metrics:")
    print(f"{'='*60}")
    print(f"Accuracy        : {acc:.4f}")
    print(f"Top-2 Accuracy  : {top2_acc:.4f}")
    print(f"Precision (Macro): {precision_macro:.4f}")
    print(f"Recall (Macro)  : {recall_macro:.4f}")
    print(f"F1-Score (Macro): {f1_macro:.4f}")
    print(f"F1-Score (Weighted): {f1_weighted:.4f}")
    print(f"AUC (Macro OVR) : {auc_macro:.4f}")
    print(f"{'='*60}")

    print(f"\nFold {fold} Classification Report:")
    print(classification_report(y_true, y_pred, target_names=CLASSES_LIST, digits=4, zero_division=0))

    # ========== SAVE EXCEL ==========
    true_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_true]
    pred_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_pred]

    df = pd.DataFrame({
        "Video_Path": video_paths,
        "True_Class": true_class_names,
        "Predicted_Class": pred_class_names,
        "Correct": [t == p for t, p in zip(true_class_names, pred_class_names)]
    })

    excel_path = os.path.join(SAVE_DIR, f"predictions_fold_{fold}.xlsx")
    df.to_excel(excel_path, index=False)
    print(f"Excel saved → {excel_path}")

    # ---------- Save model ----------
    fold_save_path = os.path.join(SAVE_DIR, f"best_hiera_activity_fold_{fold}.pth")
    torch.save({
        "model_state_dict": best_state,
        "fold": fold,
        "macro_f1": best_f1,
        "accuracy": acc,
        "top2_accuracy": top2_acc,
        "class_to_idx": full_dataset_for_split.class_to_idx,
        "idx_to_class": full_dataset_for_split.idx_to_class,
        "classes": CLASSES_LIST,
    }, fold_save_path)
    print(f"Model saved → {fold_save_path}")

    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1
        overall_best_path = fold_save_path

    del optimizer, scaler, model
    torch.cuda.empty_cache()
    fold += 1


# ============================================================
# FINAL RESULTS
# ============================================================
print("\n" + "="*70)
print("FINAL RESULTS (Hiera - 16 Classes - Last 50% Blocks + Augmentations)")
print("="*70)

print(f"\nPer-fold Accuracy : {[f'{x:.4f}' for x in accuracy_scores]}")
print(f"Mean ± std Accuracy : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")

print(f"\nPer-fold Top-2 Accuracy : {[f'{x:.4f}' for x in top2_scores]}")
print(f"Mean ± std Top-2 : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")

print(f"\nPer-fold Precision : {[f'{x:.4f}' for x in precision_scores]}")
print(f"Mean ± std Precision : {np.mean(precision_scores):.4f} ± {np.std(precision_scores):.4f}")

print(f"\nPer-fold Recall : {[f'{x:.4f}' for x in recall_scores]}")
print(f"Mean ± std Recall : {np.mean(recall_scores):.4f} ± {np.std(recall_scores):.4f}")

print(f"\nPer-fold Macro F1 : {[f'{x:.4f}' for x in f1_macro_scores]}")
print(f"Mean ± std Macro F1 : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")

print(f"\nPer-fold Weighted F1 : {[f'{x:.4f}' for x in f1_weighted_scores]}")
print(f"Mean ± std Weighted F1 : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")

print(f"\nPer-fold AUC (Macro) : {[f'{x:.4f}' for x in auc_scores]}")
print(f"Mean ± std AUC : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")

conf_sum = np.sum(confusion_matrices, axis=0)
plot_cm(conf_sum, "Aggregated Confusion Matrix (All Folds)", normalize=False)
plot_cm(conf_sum, "Aggregated Confusion Matrix (Normalized)", normalize=True)

print(f"\nOverall best model (Macro F1={overall_best_f1:.4f}) saved at:")
print(overall_best_path)
```

```python
# ============================================================
# Toddler Activity Recognition with Hiera
# 16 Classes + Last 75% Blocks Trainable + L4 Optimized
# + Class weights computed inside each fold
# + Consistent Video Augmentations (HFlip + ColorJitter + Gaussian Noise)
# + LR logging + Top-2 Accuracy
# + Edge Deployment Metrics (Memory, FLOPs, Latency, etc.)
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

!pip install -q torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
!pip install -q hiera-transformer av opencv-python tqdm scikit-learn seaborn pandas openpyxl thop

import os
import random
import time
from pathlib import Path
from collections import Counter

import av
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.transforms import functional as TF
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix,
    precision_score, recall_score, roc_auc_score, top_k_accuracy_score
)
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns
import hiera
from thop import profile, clever_format

# ────────────────────────────────────────────────
SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    torch.backends.cudnn.benchmark = True

# ===================== L4 Optimized Settings =====================
IMAGE_HEIGHT, IMAGE_WIDTH = 224, 224
SEQUENCE_LENGTH = 16
BATCH_SIZE = 4
EPOCHS = 25
LR_BACKBONE = 1e-5          # for the unfrozen blocks
LR_HEAD = 5e-5              # for the head
WEIGHT_DECAY = 0.05
NUM_FOLDS = 5
NUM_WORKERS = 4
PATIENCE = 4
PIN_MEMORY = True
USE_AMP = True
TRAIN_LAST_PERCENT = 0.75   # Last 75% of blocks

DATA_DIR = "/content/drive/MyDrive/DatasetFinal"
SAVE_DIR = "hiera_activity_checkpoints_16cls_last75pct_L4_aug"
os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 16 Classes =====================
CLASSES_LIST = [
    'Climb_Down_Stairs',
    'Climb_Up_Stairs',
    'Crawl',
    'Drink',
    'Eat',
    'Emesis',
    'Exit_Crib',
    'Fall_Down',
    'Hold_Knife',
    'Play_Slide',
    'Play_Stacking_Rings',
    'Play_Swing',
    'Play_Walker',
    'Sleep',
    'W_Sit',
    'Walk'
]
NUM_CLASSES = len(CLASSES_LIST)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 70)
print("Device:", DEVICE)
if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
    print("GPU Memory:", round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 1), "GB")
print("Classes:", CLASSES_LIST)
print(f"Number of classes: {NUM_CLASSES}")
print(f"Strategy: Last {TRAIN_LAST_PERCENT*100:.0f}% blocks trainable + Head")
print(f"AMP: {USE_AMP} | Workers: {NUM_WORKERS} | Pin Memory: {PIN_MEMORY}")
print("=" * 70)


# ============================================================
# VIDEO AUGMENTATION (consistent across all frames of a clip)
# ============================================================
class VideoAugmentation:
    def __init__(self, size=224, is_train=True):
        self.size = size
        self.is_train = is_train
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )

    def __call__(self, frames):
        """
        frames: list of numpy arrays (H, W, C) in RGB
        Returns: Tensor of shape (T, C, H, W)
        """
        # Convert to PIL
        pil_frames = [transforms.ToPILImage()(f) for f in frames]

        # ---------- Train-time augmentations (identical params for every frame) ----------
        if self.is_train:
            # 1. Horizontal Flip
            if random.random() < 0.5:
                pil_frames = [TF.hflip(f) for f in pil_frames]

            # 2. Color Jitter (same parameters for the whole clip)
            brightness = random.uniform(0.8, 1.2)
            contrast   = random.uniform(0.8, 1.2)
            saturation = random.uniform(0.8, 1.2)
            hue        = random.uniform(-0.05, 0.05)

            pil_frames = [TF.adjust_brightness(f, brightness) for f in pil_frames]
            pil_frames = [TF.adjust_contrast(f, contrast)     for f in pil_frames]
            pil_frames = [TF.adjust_saturation(f, saturation) for f in pil_frames]
            pil_frames = [TF.adjust_hue(f, hue)               for f in pil_frames]

        # Resize + ToTensor + optional Gaussian noise
        tensors = []
        for f in pil_frames:
            f = TF.resize(f, [self.size, self.size])
            t = TF.to_tensor(f)          # (C, H, W) in [0, 1]

            # 3. Gaussian Noise (train only)
            if self.is_train and random.random() < 0.3:
                noise = torch.randn_like(t) * 0.02
                t = torch.clamp(t + noise, 0.0, 1.0)

            t = self.normalize(t)
            tensors.append(t)

        return torch.stack(tensors)      # (T, C, H, W)


# ============================================================
# FRAME EXTRACTION (raw frames only)
# ============================================================
def load_video_frames_raw(video_path, num_frames=16):
    container = av.open(video_path)
    frames = []
    for frame in container.decode(video=0):
        img = frame.to_ndarray(format="rgb24")
        frames.append(img)
    container.close()

    if len(frames) == 0:
        return None

    indices = np.linspace(0, len(frames) - 1, num_frames).astype(int)
    frames = [frames[i] for i in indices]
    return frames


# ============================================================
# DATASET
# ============================================================
class ActivityDataset(Dataset):
    def __init__(self, data_dir, classes_list, num_frames=16, size=224, is_train=True):
        self.samples = []
        self.labels = []
        self.num_frames = num_frames
        self.size = size
        self.is_train = is_train
        self.class_to_idx = {cls: idx for idx, cls in enumerate(classes_list)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}
        self.augment = VideoAugmentation(size=size, is_train=is_train)

        for class_name in classes_list:
            class_path = Path(data_dir) / class_name
            if not class_path.is_dir():
                print(f"Warning: folder not found → {class_path}")
                continue

            count = 0
            for video_file in sorted(class_path.glob("*")):
                if video_file.suffix.lower() in {".mp4", ".avi", ".mov", ".mkv"}:
                    self.samples.append(str(video_file))
                    self.labels.append(self.class_to_idx[class_name])
                    count += 1
            print(f"Loaded {count} videos from '{class_name}'")

        self.labels = np.array(self.labels)
        print(f"\nTotal videos: {len(self.samples)}")
        print("Class distribution:", dict(Counter(self.labels)))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path = self.samples[idx]
        label = self.labels[idx]
        frames = load_video_frames_raw(path, self.num_frames)

        if frames is None:
            # fallback black clip
            frames = [np.zeros((self.size, self.size, 3), dtype=np.uint8)
                      for _ in range(self.num_frames)]

        video_tensor = self.augment(frames)   # (T, C, H, W)
        return video_tensor, label, path


def collate_fn(batch):
    videos, labels, paths = zip(*batch)
    videos = torch.stack(videos)
    labels = torch.tensor(labels, dtype=torch.long)
    return videos, labels, list(paths)


# ============================================================
# FREEZE EARLY BLOCKS - Train last 75%
# ============================================================
def freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT):
    # 1. Freeze everything
    for param in model.parameters():
        param.requires_grad = False

    # 2. Always train the head
    for param in model.head.parameters():
        param.requires_grad = True

    # 3. Unfreeze last N% of blocks
    if hasattr(model, "blocks"):
        blocks = model.blocks
        num_blocks = len(blocks)
        num_to_train = max(1, int(num_blocks * train_last_percent))
        start_idx = num_blocks - num_to_train

        print(f"\n[Freeze Strategy] Total blocks: {num_blocks}")
        print(f"[Freeze Strategy] Training last {num_to_train} blocks "
              f"({train_last_percent*100:.0f}%) → blocks[{start_idx}:]")

        for i in range(start_idx, num_blocks):
            for param in blocks[i].parameters():
                param.requires_grad = True
    else:
        print("[Warning] model.blocks not found. Unfreezing all parameters.")
        for param in model.parameters():
            param.requires_grad = True

    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"[Freeze Strategy] Trainable params: {trainable:,} / {total:,} "
          f"({100 * trainable / total:.1f}%)")


# ============================================================
# EDGE DEPLOYMENT METRICS
# ============================================================
def print_edge_deployment_metrics(model, device=DEVICE, input_shape=(1, 3, 16, 224, 224)):
    """
    Prints key parameters relevant for edge deployment:
    - Memory footprint (params + peak GPU memory)
    - FLOPs
    - Inference time (latency)
    - Power draw / Thermal limits (platform notes)
    - Model accuracy (placeholder – filled after CV)
    """
    print("\n" + "=" * 70)
    print("EDGE DEPLOYMENT METRICS")
    print("=" * 70)

    model.eval()
    model.to(device)

    # ---------- 1. Memory Footprint ----------
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    param_size_mb = total_params * 4 / (1024 ** 2)          # float32
    param_size_mb_fp16 = total_params * 2 / (1024 ** 2)     # fp16 estimate

    print(f"\n[1] MEMORY FOOTPRINT")
    print(f"    Total parameters      : {total_params:,}")
    print(f"    Trainable parameters  : {trainable_params:,}")
    print(f"    Model size (FP32)     : {param_size_mb:.2f} MB")
    print(f"    Model size (FP16 est.): {param_size_mb_fp16:.2f} MB")

    # Peak GPU memory during a forward pass
    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
        torch.cuda.empty_cache()
        dummy = torch.randn(*input_shape, device=device)
        with torch.no_grad():
            _ = model(dummy)
        peak_mem_mb = torch.cuda.max_memory_allocated() / (1024 ** 2)
        print(f"    Peak GPU memory (fwd) : {peak_mem_mb:.2f} MB")
    else:
        print(f"    Peak GPU memory (fwd) : N/A (CPU only)")

    # ---------- 2. FLOPs ----------
    print(f"\n[2] COMPUTATIONAL COMPLEXITY (FLOPs)")
    try:
        dummy = torch.randn(*input_shape, device=device)
        flops, params = profile(model, inputs=(dummy,), verbose=False)
        flops_str, params_str = clever_format([flops, params], "%.3f")
        print(f"    FLOPs                 : {flops_str}")
        print(f"    Params (thop)         : {params_str}")
        print(f"    FLOPs (raw)           : {flops:,.0f}")
    except Exception as e:
        print(f"    FLOPs calculation failed: {e}")

    # ---------- 3. Inference Time (Latency) ----------
    print(f"\n[3] INFERENCE TIME (Latency)")
    dummy = torch.randn(*input_shape, device=device)
    # Warm-up
    with torch.no_grad():
        for _ in range(10):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()

    # Timed runs
    n_runs = 50
    start = time.perf_counter()
    with torch.no_grad():
        for _ in range(n_runs):
            _ = model(dummy)
    if torch.cuda.is_available():
        torch.cuda.synchronize()
    end = time.perf_counter()

    avg_latency_ms = (end - start) / n_runs * 1000
    fps = 1000.0 / avg_latency_ms
    print(f"    Avg latency (batch=1) : {avg_latency_ms:.2f} ms")
    print(f"    Throughput            : {fps:.1f} FPS")

    # ---------- 4. Power Draw ----------
    print(f"\n[4] POWER DRAW")
    print(f"    Direct measurement    : Not available in Colab / generic GPU env")
    print(f"    Notes                 : Measure on target edge device (e.g. Jetson,")
    print(f"                            Raspberry Pi, mobile NPU) using:")
    print(f"                            - tegrastats / jetson-stats (NVIDIA)")
    print(f"                            - powerstat / RAPL (Intel)")
    print(f"                            - battery / board-level sensors")
    print(f"    Typical range (est.)  : 5–15 W for similar video models on edge GPUs")

    # ---------- 5. Thermal Limits ----------
    print(f"\n[5] THERMAL LIMITS")
    print(f"    Direct measurement    : Not available in Colab")
    print(f"    Notes                 : Monitor on-device temperature under sustained load.")
    print(f"                            Typical edge limits:")
    print(f"                            - Mobile SoC / NPU : 70–85 °C")
    print(f"                            - Jetson / embedded GPU : 80–95 °C")
    print(f"                            - Sustained thermal throttling reduces FPS")

    # ---------- 6. Model Accuracy ----------
    print(f"\n[6] MODEL ACCURACY")
    print(f"    (Filled after cross-validation completes)")
    print(f"    Target metrics        : Accuracy, Top-2 Acc, Macro-F1, Weighted-F1, AUC")
    print(f"    See FINAL RESULTS section below for measured values.")

    print("=" * 70 + "\n")


# ============================================================
# LOAD FULL DATASET (labels only – we create train/val subsets later)
# ============================================================
print("\nLoading full dataset (for labels & paths)...")
# We only need the labels and paths; actual tensors are created on-the-fly
# with different augmentations for train vs val.
full_dataset_for_split = ActivityDataset(
    DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
)


# ============================================================
# MODEL
# ============================================================
def create_hiera_model():
    model = hiera.hiera_base_16x224(
        pretrained=True,
        checkpoint="mae_k400_ft_k400",
        num_classes=400,
    )
    in_features = model.head.projection.in_features
    model.head.projection = nn.Linear(in_features, NUM_CLASSES)

    # Unfreeze last 75% blocks + head
    freeze_early_blocks(model, train_last_percent=TRAIN_LAST_PERCENT)

    return model.to(DEVICE)


# ============================================================
# TRAIN / EVAL (with AMP + Top-2)
# ============================================================
def train_one_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    total_loss, correct, total = 0.0, 0, 0

    for videos, labels, _ in tqdm(loader, desc=" train", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()   # (B, C, T, H, W)

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * videos.size(0)
        preds = logits.argmax(1)
        correct += (preds == labels).sum().item()
        total += videos.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    all_preds, all_labels, all_probs, all_paths = [], [], [], []

    for videos, labels, paths in tqdm(loader, desc=" val", leave=False):
        videos = videos.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        videos = videos.permute(0, 2, 1, 3, 4).contiguous()

        with torch.amp.autocast('cuda', enabled=USE_AMP):
            logits = model(videos)
            loss = criterion(logits, labels)

        total_loss += loss.item() * videos.size(0)
        probs = torch.softmax(logits.float(), dim=1)
        preds = logits.argmax(1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
        all_paths.extend(paths)

    return (total_loss / len(loader.dataset),
            np.array(all_preds),
            np.array(all_labels),
            np.array(all_probs),
            all_paths)


def plot_cm(cm, title, normalize=True):
    if normalize:
        cm_plot = cm.astype("float") / (cm.sum(axis=1)[:, np.newaxis] + 1e-8)
        fmt = ".2f"
    else:
        cm_plot = cm
        fmt = "d"

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm_plot, annot=True, fmt=fmt, cmap="Blues",
                xticklabels=CLASSES_LIST, yticklabels=CLASSES_LIST)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# PROFILE MODEL ONCE (before CV) for Edge Metrics
# ============================================================
print("\nCreating temporary model for edge deployment profiling...")
_temp_model = create_hiera_model()
print_edge_deployment_metrics(_temp_model)
del _temp_model
torch.cuda.empty_cache()


# ============================================================
# 5-FOLD STRATIFIED CROSS VALIDATION
# ============================================================
skf = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=SEED)

accuracy_scores = []
top2_scores = []
f1_macro_scores = []
f1_weighted_scores = []
precision_scores = []
recall_scores = []
auc_scores = []
confusion_matrices = []

overall_best_f1 = 0.0
overall_best_path = None

fold = 1
for train_idx, val_idx in skf.split(np.zeros(len(full_dataset_for_split)), full_dataset_for_split.labels):
    print(f"\n{'='*70}")
    print(f"========== Fold {fold}/{NUM_FOLDS} ==========")
    print(f"{'='*70}")

    # ---------- Create train / val datasets with correct augmentation ----------
    train_dataset = ActivityDataset(
        DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=True
    )
    val_dataset = ActivityDataset(
        DATA_DIR, CLASSES_LIST, SEQUENCE_LENGTH, IMAGE_HEIGHT, is_train=False
    )

    train_subset = Subset(train_dataset, train_idx)
    val_subset   = Subset(val_dataset,   val_idx)

    # ---------- Compute class weights using ONLY training labels ----------
    train_labels = full_dataset_for_split.labels[train_idx]
    class_weights = compute_class_weight(
        "balanced",
        classes=np.unique(train_labels),
        y=train_labels
    )
    class_weight_tensor = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
    print("Class weights (this fold):", dict(enumerate(np.round(class_weights, 3))))

    train_loader = DataLoader(
        train_subset, batch_size=BATCH_SIZE, shuffle=True,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )
    val_loader = DataLoader(
        val_subset, batch_size=BATCH_SIZE, shuffle=False,
        num_workers=NUM_WORKERS, collate_fn=collate_fn,
        pin_memory=PIN_MEMORY, persistent_workers=True if NUM_WORKERS > 0 else False
    )

    model = create_hiera_model()
    criterion = nn.CrossEntropyLoss(weight=class_weight_tensor)

    # Differential learning rates
    backbone_params = [p for n, p in model.named_parameters() if "head" not in n and p.requires_grad]
    head_params     = [p for n, p in model.named_parameters() if "head" in n and p.requires_grad]

    optimizer = torch.optim.AdamW([
        {"params": backbone_params, "lr": LR_BACKBONE},
        {"params": head_params,     "lr": LR_HEAD}
    ], weight_decay=WEIGHT_DECAY)

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP)

    best_f1 = 0.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(EPOCHS):
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler)
        val_loss, y_pred, y_true, y_prob, _ = evaluate(model, val_loader, criterion)
        scheduler.step()

        # ---------- LR logging ----------
        current_lrs = [group['lr'] for group in optimizer.param_groups]
        print(f"  → LRs: backbone={current_lrs[0]:.2e} | head={current_lrs[1]:.2e}")

        val_acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average="macro", zero_division=0)

        # Top-2 accuracy
        try:
            top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
        except Exception:
            top2_acc = 0.0

        print(f"Epoch {epoch+1:02d} | "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} "
              f"top2={top2_acc:.4f} macro_F1={f1_macro:.4f}")

        if f1_macro > best_f1:
            best_f1 = f1_macro
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
            print(f" → New best macro F1: {best_f1:.4f}")
        else:
            epochs_no_improve += 1
            print(f" → No improvement for {epochs_no_improve}/{PATIENCE} epochs")

        if epochs_no_improve >= PATIENCE:
            print(f"Early stopping at epoch {epoch+1}")
            break

    # ---------- Final evaluation with best weights ----------
    model.load_state_dict(best_state)
    model.to(DEVICE)
    val_loss, y_pred, y_true, y_prob, video_paths = evaluate(model, val_loader, criterion)

    acc = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall_macro    = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1_macro        = f1_score(y_true, y_pred, average="macro", zero_division=0)
    f1_weighted     = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    try:
        top2_acc = top_k_accuracy_score(y_true, y_prob, k=2, labels=list(range(NUM_CLASSES)))
    except Exception:
        top2_acc = 0.0

    try:
        y_true_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        auc_macro = roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr")
    except Exception as e:
        print(f"AUC calculation warning: {e}")
        auc_macro = 0.0

    accuracy_scores.append(acc)
    top2_scores.append(top2_acc)
    f1_macro_scores.append(f1_macro)
    f1_weighted_scores.append(f1_weighted)
    precision_scores.append(precision_macro)
    recall_scores.append(recall_macro)
    auc_scores.append(auc_macro)

    cm = confusion_matrix(y_true, y_pred)
    confusion_matrices.append(cm)
    plot_cm(cm, f"Activity Recognition CM – Fold {fold}")

    print(f"\n{'='*60}")
    print(f"Fold {fold} Final Metrics:")
    print(f"{'='*60}")
    print(f"Accuracy        : {acc:.4f}")
    print(f"Top-2 Accuracy  : {top2_acc:.4f}")
    print(f"Precision (Macro): {precision_macro:.4f}")
    print(f"Recall (Macro)  : {recall_macro:.4f}")
    print(f"F1-Score (Macro): {f1_macro:.4f}")
    print(f"F1-Score (Weighted): {f1_weighted:.4f}")
    print(f"AUC (Macro OVR) : {auc_macro:.4f}")
    print(f"{'='*60}")

    print(f"\nFold {fold} Classification Report:")
    print(classification_report(y_true, y_pred, target_names=CLASSES_LIST, digits=4, zero_division=0))

    # ========== SAVE EXCEL ==========
    true_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_true]
    pred_class_names = [full_dataset_for_split.idx_to_class[i] for i in y_pred]

    df = pd.DataFrame({
        "Video_Path": video_paths,
        "True_Class": true_class_names,
        "Predicted_Class": pred_class_names,
        "Correct": [t == p for t, p in zip(true_class_names, pred_class_names)]
    })

    excel_path = os.path.join(SAVE_DIR, f"predictions_fold_{fold}.xlsx")
    df.to_excel(excel_path, index=False)
    print(f"Excel saved → {excel_path}")

    # ---------- Save model ----------
    fold_save_path = os.path.join(SAVE_DIR, f"best_hiera_activity_fold_{fold}.pth")
    torch.save({
        "model_state_dict": best_state,
        "fold": fold,
        "macro_f1": best_f1,
        "accuracy": acc,
        "top2_accuracy": top2_acc,
        "class_to_idx": full_dataset_for_split.class_to_idx,
        "idx_to_class": full_dataset_for_split.idx_to_class,
        "classes": CLASSES_LIST,
    }, fold_save_path)
    print(f"Model saved → {fold_save_path}")

    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1
        overall_best_path = fold_save_path

    del optimizer, scaler, model
    torch.cuda.empty_cache()
    fold += 1


# ============================================================
# FINAL RESULTS
# ============================================================
print("\n" + "="*70)
print("FINAL RESULTS (Hiera - 16 Classes - Last 75% Blocks + Augmentations)")
print("="*70)

print(f"\nPer-fold Accuracy : {[f'{x:.4f}' for x in accuracy_scores]}")
print(f"Mean ± std Accuracy : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")

print(f"\nPer-fold Top-2 Accuracy : {[f'{x:.4f}' for x in top2_scores]}")
print(f"Mean ± std Top-2 : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")

print(f"\nPer-fold Precision : {[f'{x:.4f}' for x in precision_scores]}")
print(f"Mean ± std Precision : {np.mean(precision_scores):.4f} ± {np.std(precision_scores):.4f}")

print(f"\nPer-fold Recall : {[f'{x:.4f}' for x in recall_scores]}")
print(f"Mean ± std Recall : {np.mean(recall_scores):.4f} ± {np.std(recall_scores):.4f}")

print(f"\nPer-fold Macro F1 : {[f'{x:.4f}' for x in f1_macro_scores]}")
print(f"Mean ± std Macro F1 : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")

print(f"\nPer-fold Weighted F1 : {[f'{x:.4f}' for x in f1_weighted_scores]}")
print(f"Mean ± std Weighted F1 : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")

print(f"\nPer-fold AUC (Macro) : {[f'{x:.4f}' for x in auc_scores]}")
print(f"Mean ± std AUC : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")

conf_sum = np.sum(confusion_matrices, axis=0)
plot_cm(conf_sum, "Aggregated Confusion Matrix (All Folds)", normalize=False)
plot_cm(conf_sum, "Aggregated Confusion Matrix (Normalized)", normalize=True)

print(f"\nOverall best model (Macro F1={overall_best_f1:.4f}) saved at:")
print(overall_best_path)

# ============================================================
# EDGE DEPLOYMENT SUMMARY (with final accuracy)
# ============================================================
print("\n" + "=" * 70)
print("EDGE DEPLOYMENT SUMMARY (with measured accuracy)")
print("=" * 70)
print(f"Memory Footprint     : See earlier EDGE DEPLOYMENT METRICS section")
print(f"FLOPs                : See earlier EDGE DEPLOYMENT METRICS section")
print(f"Inference Time       : See earlier EDGE DEPLOYMENT METRICS section")
print(f"Power Draw           : Measure on target edge hardware")
print(f"Thermal Limits       : Monitor on target edge hardware")
print(f"Model Accuracy       : {np.mean(accuracy_scores):.4f} ± {np.std(accuracy_scores):.4f}")
print(f"Top-2 Accuracy       : {np.mean(top2_scores):.4f} ± {np.std(top2_scores):.4f}")
print(f"Macro F1             : {np.mean(f1_macro_scores):.4f} ± {np.std(f1_macro_scores):.4f}")
print(f"Weighted F1          : {np.mean(f1_weighted_scores):.4f} ± {np.std(f1_weighted_scores):.4f}")
print(f"AUC (Macro)          : {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("=" * 70)
```
